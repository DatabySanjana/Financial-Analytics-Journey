```python
# Simple Interest
# A company invests ₹2,50,000 at 8% simple interest per year for 4 years. Calculate: Simple Interest & Final Amount
principal = 250000
rate = 0.08
time = 4
# Simple Interest
simple_interest = principal * rate * time
# Final Amount
final_amount = principal + simple_interest
print("Simple Interest:", simple_interest)
print("Final Amount:", final_amount)
```

```python
# Find the Principal
# An investment earns ₹36,000 in simple interest at 9% per year over 5 years. Calculate: Principal
simple_interest = 36000
rate = 0.09
time = 5
# Principal
principal = simple_interest / (rate * time)
print("Principal:", principal)
```

```python
# Compound Interest
# A company invests ₹3,00,000 at 10% annual compound interest for 4 years. Calculate: Future Value & Compound Interest Earned
principal = 300000
rate = 0.10
time = 4
# Future Value
future_value = principal * (1 + rate) ** time
# Compound Interest
compound_interest = future_value - principal
print("Future Value:", future_value)
print("Compound Interest:", compound_interest)
```

```python
# Simple Interest vs Compound Interest
# An investor invests ₹5,00,000 at 8% annually for 5 years.
principal = 500000
rate = 0.08
time = 5
# Simple Interest Amount
simple_amount = principal * (1 + rate * time)
# Compound Interest Amount
compound_amount = principal * (1 + rate) ** time
# Difference
difference = compound_amount - simple_amount
print("Simple Interest Amount:", simple_amount)
print("Compound Interest Amount:", compound_amount)
print("Difference:", difference)
```

```python
# Future Value(FV)
# A company invests ₹4,00,000 today at an annual return of 9% for 6 years. Calculate: Future Value
present_value = 400000
rate = 0.09
time = 6
# Future Value
future_value = present_value * (1 + rate) ** time
print("Future Value:", future_value)
```

```python
# Present Value(PV)
# A business expects to receive ₹10,00,000 after 5 years. Required return / discount rate = 10%. Calculate: Present Value
future_value = 1000000
discount_rate = 0.10
time = 5
# Present Value
present_value = future_value / (1 + discount_rate) ** time
print("Present Value:", present_value)
```

```python
# Discounting Multiple Cash Flows
# A project is expected to generate: Year1 = ₹200000, Year2 = ₹250000, Year3 = ₹300000, Year4 = ₹350000, Discount rate = 10%
# Calculate:
# 1. PV of each cash flow
# 2. Total Present Value
cash_flow_1 = 200000
cash_flow_2 = 250000
cash_flow_3 = 300000
cash_flow_4 = 350000
discount_rate = 0.10
# PV of each cash flow
pv_1 = cash_flow_1 / (1 + discount_rate) ** 1
pv_2 = cash_flow_2 / (1 + discount_rate) ** 2
pv_3 = cash_flow_3 / (1 + discount_rate) ** 3
pv_4 = cash_flow_4 / (1 + discount_rate) ** 4
# Total Present Value
total_pv = pv_1 + pv_2 + pv_3 + pv_4
print("PV Year 1:", pv_1)
print("PV Year 2:", pv_2)
print("PV Year 3:", pv_3)
print("PV Year 4:", pv_4)
print("Total Present Value:", total_pv)
```

```python
# Inflation-Adjusted Value
# A company has ₹10,00,000 today. Inflation = 6% per year & Time = 8 years
# Calculate: Purchasing-power value after 8 years & Interpret the result
current_value = 1000000
inflation_rate = 0.06
time = 8
# Inflation-adjusted value
adjusted_value = current_value / (1 + inflation_rate) ** time
print("Inflation-Adjusted Value:", adjusted_value)
# Interpretation
print("Interpretation: The purchasing power of ₹1000000")
print("will fall to approximately ₹6.27 lakh in today's money.")
```

```python
# Required Investment Today
# A company wants to have ₹25,00,000 after 7 years. Expected annual return = 8%.
# Investment required today
future_value = 2500000
rate = 0.08
time = 7
# Present Value / Investment Today
investment_today = future_value / (1 + rate) ** time
print("Investment Required Today:", investment_today)
```

```python
# CAGR
# A company's revenue increased from ₹40 lakh to ₹75 lakh over 5 years. Calculate: CAGR & Interpret the result
initial_value = 40
final_value = 75
time = 5
# CAGR
cagr = (final_value / initial_value) ** (1 / time) - 1
print("CAGR:", cagr * 100, "%")
# Interpretation
print("Interpretation: Revenue grew at approximately",
      round(cagr * 100, 2),
      "% per year on a compounded basis.")
```

```python
# Investment CAGR Comparison
# Investment A: Initial Value = ₹500000, Final Value = ₹800000, Period = 4 years
# Investment B: Initial Value = ₹1000000, Final Value = ₹1600000, Period = 6 years
# Calculate: CAGR of Investment A, CAGR of Investment B, Identify the investment with higher CAGR
initial_A = 500000
final_A = 800000
time_A = 4
initial_B = 1000000
final_B = 1600000
time_B = 6
# CAGR of Investment A
cagr_A = (final_A / initial_A) ** (1 / time_A) - 1
# CAGR of Investment B
cagr_B = (final_B / initial_B) ** (1 / time_B) - 1
print("Investment A CAGR:", cagr_A * 100, "%")
print("Investment B CAGR:", cagr_B * 100, "%")
# Higher CAGR
if cagr_A > cagr_B:
    print("Higher CAGR: Investment A")
else:
    print("Higher CAGR: Investment B")
```

```python
# Annuity
# A company expects to receive ₹2,00,000 at the end of every year for 5 years. Discount rate = 10%. Calculate: Present Value of the annuity
payment = 200000
rate = 0.10
time = 5
# Present Value of Annuity
annuity_pv = payment * (1 - (1 + rate) ** -time) / rate
print("Present Value of Annuity:", annuity_pv)
```

```python
# EMI
# Loan Principal = ₹800000, Annual Interest Rate = 9%, Tenure = 5 years. Calculate: Monthly EMI, Total Amount Paid, Total Interest Paid
principal = 800000
annual_rate = 0.09
years = 5
monthly_rate = annual_rate / 12
months = years * 12
# Monthly EMI
emi = principal * monthly_rate * (1 + monthly_rate) ** months / \
      ((1 + monthly_rate) ** months - 1)
# Total Amount Paid
total_paid = emi * months
# Total Interest
total_interest = total_paid - principal
print("Monthly EMI:", emi)
print("Total Amount Paid:", total_paid)
print("Total Interest Paid:", total_interest)
```

```python
# Q14 — Loan Comparison
# Loan A: Principal = ₹1000000, Interest = 9%, Tenure = 5 years
# Loan B: Principal = ₹1000000, Interest = 8%, Tenure = 7 years
# Calculate for both: Monthly EMI, Total Repayment, Total Interest
# Then identify: Lower EMI & Lower Total Interest
principal = 1000000
# Loan A
rate_A = 0.09
years_A = 5
months_A = years_A * 12
monthly_rate_A = rate_A / 12
emi_A = principal * monthly_rate_A * (1 + monthly_rate_A) ** months_A / \
        ((1 + monthly_rate_A) ** months_A - 1)
total_A = emi_A * months_A
interest_A = total_A - principal
# Loan B
rate_B = 0.08
years_B = 7
months_B = years_B * 12
monthly_rate_B = rate_B / 12
emi_B = principal * monthly_rate_B * (1 + monthly_rate_B) ** months_B / \
        ((1 + monthly_rate_B) ** months_B - 1)
total_B = emi_B * months_B
interest_B = total_B - principal
print("Loan A EMI:", emi_A)
print("Loan A Total Repayment:", total_A)
print("Loan A Total Interest:", interest_A)
print("Loan B EMI:", emi_B)
print("Loan B Total Repayment:", total_B)
print("Loan B Total Interest:", interest_B)
if emi_A < emi_B:
    print("Lower EMI: Loan A")
else:
    print("Lower EMI: Loan B")
if interest_A < interest_B:
    print("Lower Total Interest: Loan A")
else:
    print("Lower Total Interest: Loan B")
```

```python
# Loan Amortization
# Using Loan A from above:Principal = ₹1000000, Interest = 9%, Tenure = 5 years / 60 months
# Create an amortization table containing: Month, Opening Balance, EMI, Interest, Principal, Closing Balance
# Then calculate: Total Interest, Total Principal, Year 1 Interest, Year 1 Principal, Outstanding Balance after Year 3
import pandas as pd
principal = 1000000
annual_rate = 0.09
months = 60
monthly_rate = annual_rate / 12
emi = principal * monthly_rate * (1 + monthly_rate) ** months / \
      ((1 + monthly_rate) ** months - 1)
opening_balance = principal
amortization_data = []
for month in range(1, months + 1):
# Interest for current month
    interest = opening_balance * monthly_rate
# Principal portion of EMI
    principal_paid = emi - interest
# Remaining loan balance
    closing_balance = opening_balance - principal_paid
    amortization_data.append([
        month,
        opening_balance,
        emi,
        interest,
        principal_paid,
        closing_balance])
    opening_balance = closing_balance
amortization = pd.DataFrame(
    amortization_data,
    columns=[
        "Month",
        "Opening Balance",
        "EMI",
        "Interest",
        "Principal",
        "Closing Balance"])
amortization = amortization.round(2)
print(amortization)
# Total Interest
total_interest = amortization["Interest"].sum()
# Total Principal
total_principal = amortization["Principal"].sum()
# Year 1 Interest
year_1_interest = amortization.iloc[:12]["Interest"].sum()
# Year 1 Principal
year_1_principal = amortization.iloc[:12]["Principal"].sum()
# Outstanding Balance after Year 3
balance_after_year_3 = amortization.iloc[35]["Closing Balance"]
print("Total Interest:", total_interest)
print("Total Principal:", total_principal)
print("Year 1 Interest:", year_1_interest)
print("Year 1 Principal:", year_1_principal)
print("Outstanding Balance after Year 3:", balance_after_year_3)
```

```python
# ROI Comparison
# Investment A: Initial Investment = ₹500000, Final Value = ₹650000
# Investment B: Initial Investment = ₹800000, Final Value = ₹1000000
# Investment C: Initial Investment = ₹1200000, Final Value = ₹1560000
# Calculate: ROI for A,B,C & Identify the investment with highest ROI.
investment_A = 500000
final_A = 650000
investment_B = 800000
final_B = 1000000
investment_C = 1200000
final_C = 1560000
# ROI A
roi_A = (final_A - investment_A) / investment_A * 100
# ROI B
roi_B = (final_B - investment_B) / investment_B * 100
# ROI C
roi_C = (final_C - investment_C) / investment_C * 100
print("Investment A ROI:", roi_A, "%")
print("Investment B ROI:", roi_B, "%")
print("Investment C ROI:", roi_C, "%")
# Highest ROI
if roi_A >= roi_B and roi_A >= roi_C:
    print("Highest ROI: Investment A")
elif roi_B >= roi_A and roi_B >= roi_C:
    print("Highest ROI: Investment B")
else:
    print("Highest ROI: Investment C")
```

```python
# Payback Period
# Initial Investment = ₹1000000, Annual Cash Inflows: Year1 = ₹200000, Year2 = ₹250000, Year3 = ₹300000, Year4 = ₹350000, Year5 = ₹200000
# Calculate: Cumulative cash flows, Amount remaining after Year 3, Fraction of Year 4 required, Payback Period.
initial_investment = 1000000
cash_flow_1 = 200000
cash_flow_2 = 250000
cash_flow_3 = 300000
cash_flow_4 = 350000
cash_flow_5 = 200000
# Cumulative cash flows
cumulative_1 = cash_flow_1
cumulative_2 = cumulative_1 + cash_flow_2
cumulative_3 = cumulative_2 + cash_flow_3
cumulative_4 = cumulative_3 + cash_flow_4
# Amount remaining after Year 3
remaining_amount = initial_investment - cumulative_3
# Fraction of Year 4 required
fraction_of_year = remaining_amount / cash_flow_4
# Payback Period
payback_period = 3 + fraction_of_year
print("Payback Period:", payback_period, "years")
```

```python
# Project Profitability
# Initial Investment = ₹1500000, Cash Flows: Year1 = ₹400000, Year2 = ₹450000, Year3 = ₹500000, Year4 = ₹550000, Year5 = ₹600000
# Required Return = 10%. Calculate: Total Cash Inflows, ROI, Payback Period, NPV, IRR, Investment Decision. 
import numpy as np
initial_investment = 1500000
cash_flows = np.array([400000, 450000, 500000, 550000, 600000])
discount_rate = 0.10
# Total Cash Inflows
total_cash_inflows = np.sum(cash_flows)
# ROI
roi = (total_cash_inflows - initial_investment) / initial_investment * 100
# Payback Period
cumulative_cash_flow = np.cumsum(cash_flows)
remaining_amount = initial_investment - cumulative_cash_flow[2]
fraction_of_year = remaining_amount / cash_flows[3]
payback_period = 3 + fraction_of_year
# NPV
years = np.arange(1, 6)
present_values = cash_flows / (1 + discount_rate) ** years
npv = np.sum(present_values) - initial_investment
# IRR ≈ 18.34%
irr = 0.1834
# Investment Decision
if npv > 0 and irr > discount_rate:
    decision = "ACCEPT"
else:
    decision = "REJECT"
print("Total Cash Inflows:", total_cash_inflows)
print("ROI:", round(roi, 2), "%")
print("Payback Period:", round(payback_period, 2), "years")
print("NPV:", round(npv, 2))
print("IRR:", round(irr * 100, 2), "%")
print("Investment Decision:", decision)
```

```python
# Investment Comparison
# Project A: Initial Investment = ₹2000000
# Cash Flows: Year1 = ₹500000, Year2 = ₹600000, Year3 = ₹700000, Year4 = ₹800000, Year5 = ₹900000
# Project B: Initial Investment = ₹2000000
# Cash Flows: Year1 = ₹200000, Year2 = ₹500000, Year3 = ₹800000, Year4 = ₹1000000, Year5 = ₹1200000
# Discount Rate = 10%. Calculate: Total Cash Flow, ROI, NPV, IRR, Payback Period & Compare both projects.
import numpy as np
import pandas as pd
initial_investment = 2000000
discount_rate = 0.10
project_A = np.array([500000, 600000, 700000, 800000, 900000])
project_B = np.array([200000, 500000, 800000, 1000000, 1200000])
# PROJECT A
# Total Cash Flow
total_cash_flow_A = np.sum(project_A)
# ROI
roi_A = (total_cash_flow_A - initial_investment) / initial_investment * 100
# NPV
years = np.arange(1, 6)
pv_A = project_A / (1 + discount_rate) ** years
npv_A = np.sum(pv_A) - initial_investment
# IRR
irr_A = 0.1971
# Payback Period
cumulative_A = np.cumsum(project_A)
remaining_A = initial_investment - cumulative_A[2]
fraction_A = remaining_A / project_A[3]
payback_A = 3 + fraction_A
# PROJECT B
# Total Cash Flow
total_cash_flow_B = np.sum(project_B)
# ROI
roi_B = (total_cash_flow_B - initial_investment) / initial_investment * 100
# NPV
pv_B = project_B / (1 + discount_rate) ** years
npv_B = np.sum(pv_B) - initial_investment
# IRR
irr_B = 0.1895
# Payback Period
cumulative_B = np.cumsum(project_B)
remaining_B = initial_investment - cumulative_B[2]
fraction_B = remaining_B / project_B[3]
payback_B = 3 + fraction_B
# COMPARISON TABLE
comparison = pd.DataFrame({
    "Project A": [
        total_cash_flow_A,
        roi_A,
        npv_A,
        irr_A * 100,
        payback_A],
    "Project B": [
        total_cash_flow_B,
        roi_B,
        npv_B,
        irr_B * 100,
        payback_B]
}, index=[
    "Total Cash Flow",
    "ROI (%)",
    "NPV",
    "IRR (%)",
    "Payback Period"])
print(comparison)
```

```python
# Financing Decision
# Company needs ₹25 lakh.
# Option A: Interest = 9%, Tenure = 5 years
# Option B: Interest = 8%, Tenure = 7 years
# Calculate: EMI for both, Total Repayment, Total Interest, Best option for lower monthly cash-flow pressure, Best option for lower total financing cost
principal = 2500000
# Option A
rate_A = 0.09
years_A = 5
months_A = years_A * 12
monthly_rate_A = rate_A / 12
emi_A = principal * monthly_rate_A * (1 + monthly_rate_A) ** months_A / \
        ((1 + monthly_rate_A) ** months_A - 1)
total_A = emi_A * months_A
interest_A = total_A - principal
# Option B
rate_B = 0.08
years_B = 7
months_B = years_B * 12
monthly_rate_B = rate_B / 12
emi_B = principal * monthly_rate_B * (1 + monthly_rate_B) ** months_B / \
        ((1 + monthly_rate_B) ** months_B - 1)
total_B = emi_B * months_B
interest_B = total_B - principal
print("Option A EMI:", emi_A)
print("Option A Total Repayment:", total_A)
print("Option A Total Interest:", interest_A)
print("Option B EMI:", emi_B)
print("Option B Total Repayment:", total_B)
print("Option B Total Interest:", interest_B)
if emi_A < emi_B:
    print("Lower Monthly Cash-Flow Pressure: Option A")
else:
    print("Lower Monthly Cash-Flow Pressure: Option B")
if interest_A < interest_B:
    print("Lower Total Financing Cost: Option A")
else:
    print("Lower Total Financing Cost: Option B")
```

```python
# Final FP&A Case
# Company has ₹3000000 available for investment.
# Project A — Expansion: Initial Investment = ₹3000000
# Cash Flows: Year1 = ₹700000, Year2 = ₹800000, Year3 = ₹900000, Year4 = ₹1000000, Year5 = ₹1100000
# Project B — Technology Upgrade: Initial Investment = ₹3000000
# Cash Flows: Year1 = ₹300000, Year2 = ₹700000, Year3 = ₹1000000, Year4 = ₹1300000, Year5 = ₹1500000, Required Return = 10%
# Calculate: Total Cash Flow, ROI, NPV, IRR, Payback Period
# Basic information
initial_investment = 3000000
discount_rate = 0.10
project_A = np.array([700000, 800000, 900000, 1000000, 1100000])
project_B = np.array([300000, 700000, 1000000, 1300000, 1500000])
years = np.arange(1, 6)
# PROJECT A
# Total Cash Flow
total_cash_flow_A = np.sum(project_A)
# ROI
roi_A = (
    (total_cash_flow_A - initial_investment)
    / initial_investment
) * 100
# NPV
pv_A = project_A / (1 + discount_rate) ** years
npv_A = np.sum(pv_A) - initial_investment
# Payback Period
cumulative_A = np.cumsum(project_A)
for year, cumulative in enumerate(cumulative_A, start=1):
    if cumulative >= initial_investment:
        previous_cumulative = cumulative - project_A[year - 1]
        remaining_amount = initial_investment - previous_cumulative
        fraction = remaining_amount / project_A[year - 1]
        payback_A = (year - 1) + fraction
        break
# IRR
cash_flow_A_with_investment = np.insert(project_A, 0, -initial_investment)
rates = np.arange(0, 1, 0.0001)
npv_values_A = []
for rate in rates:
    npv_value = np.sum(
        cash_flow_A_with_investment
        / (1 + rate) ** np.arange(0, 6))
    npv_values_A.append(npv_value)
closest_index_A = np.argmin(np.abs(npv_values_A))
irr_A = rates[closest_index_A]
# PROJECT B
# Total Cash Flow
total_cash_flow_B = np.sum(project_B)
# ROI
roi_B = (
    (total_cash_flow_B - initial_investment)
    / initial_investment
) * 100
# NPV
pv_B = project_B / (1 + discount_rate) ** years
npv_B = np.sum(pv_B) - initial_investment
# Payback Period
cumulative_B = np.cumsum(project_B)
for year, cumulative in enumerate(cumulative_B, start=1):
    if cumulative >= initial_investment:
        previous_cumulative = cumulative - project_B[year - 1]
        remaining_amount = initial_investment - previous_cumulative
        fraction = remaining_amount / project_B[year - 1]
        payback_B = (year - 1) + fraction
        break
# IRR
cash_flow_B_with_investment = np.insert(project_B, 0, -initial_investment)
npv_values_B = []
for rate in rates:
    npv_value = np.sum(
        cash_flow_B_with_investment
        / (1 + rate) ** np.arange(0, 6))
    npv_values_B.append(npv_value)
closest_index_B = np.argmin(np.abs(npv_values_B))
irr_B = rates[closest_index_B]
# COMPARISON TABLE
comparison = pd.DataFrame({
    "Project A": [
        total_cash_flow_A,
        roi_A,
        npv_A,
        irr_A * 100,
        payback_A],
    "Project B": [
        total_cash_flow_B,
        roi_B,
        npv_B,
        irr_B * 100,
        payback_B]
}, index=[
    "Total Cash Flow",
    "ROI (%)",
    "NPV",
    "IRR (%)",
    "Payback Period (Years)"])
print(comparison.round(2))
```

#### Learned how different financial metrics evaluate investments from different perspectives: ROI measures overall return, NPV measures value created in today's terms, IRR measures the investment's implied return and Payback Period measures how quickly the initial investment is recovered.
