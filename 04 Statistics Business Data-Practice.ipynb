### Level 1 - Fundamentals

```python
# Population vs Sample
# Question: A company has 50000 customers. An analyst randomly selects 1000 customers to study their spending.
# Identify the population and sample and explain why a sample might be used.
population = 50_000
sample = 1_000
print("Population:", population, "customers")
print("Sample:", sample, "customers")
print("Reason: A sample saves time, cost, and effort.")
```

```python
# Mean
# Question: Calculate the mean: 12, 15, 18, 20, 25, 30
data = [12, 15, 18, 20, 25, 30]
mean = sum(data) / len(data)
print("Mean:", mean)
```

```python
# Median
# Question: Calculate the median: 14, 18, 21, 25, 30, 35, 40
data = [14, 18, 21, 25, 30, 35, 40]
data.sort()
n = len(data)
median = data[n // 2]
print("Median:", median)
```

```python
# Mode
# Question: Find the mode: 5, 7, 8, 7, 9, 10, 7, 8, 12
from collections import Counter
data = [5, 7, 8, 7, 9, 10, 7, 8, 12]
frequency = Counter(data)
mode = frequency.most_common(1)[0][0]
print("Mode: ", mode)
```

```python
# Range
# Question: Calculate the range: 25, 32, 18, 45, 29, 51, 37
data = [25, 32, 18, 45, 29, 51, 37]
data_range = max(data) - min(data)
print("Range: ", data_range)
```

#### Level 2: Quartiles, IQR & Percentile

```python
# Calculate Q1, Q2 and Q3
import numpy as np
data = [10, 15, 20, 25, 30, 35, 40, 45, 50]
Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)
print("Data:", data)
print("Q1:", Q1)
print("Q2 (Median):", Q2)
print("Q3:", Q3)
```

```python
# Calculate IQR
IQR = Q3 - Q1
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Percentile
# Question: A student scored in the 85th percentile. Explain what this means.
percentile = 85
print(
    f"85th percentile: The student scored higher than "
    f"approximately {percentile}% of students.")
```

```python
# Business Percentile
# Question: A customer's monthly spending is at the 90th percentile. What does this tell us?
percentile = 90
print(
    f"Customer spending: Higher than approximately "
    f"{percentile}% of customers.")
print("Interpretation: The customer is a high-spending customer.")
```

### Level 3 - Variance & Standard Deviation

```python
# Population Variance
data = [2, 4, 6, 8, 10]
mean = np.mean(data)
variance = np.var(data)
print("Data:", data)
print("Mean:", mean)
print("Population Variance:", variance)
```

```python
# Population Standard Deviation
# Question: Calculate the population standard deviation.
data = [2, 4, 6, 8, 10]
standard_deviation = np.std(data)
print("Population Standard Deviation: ", round(standard_deviation, 2))
```

```python
# Compare Variability
business_A = [98, 99, 100, 101, 102]
business_B = [70, 85, 100, 115, 130]
sd_A = np.std(business_A)
sd_B = np.std(business_B)
print("Business A")
print("Mean:", np.mean(business_A))
print("Standard Deviation:", sd_A)

print("\nBusiness B")
print("Mean:", np.mean(business_B))
print("Standard Deviation:", sd_B)

print("Higher variability: ", "Business A" if sd_A > sd_B else "Business B")
print("More volatile: ", "Business A" if sd_A > sd_B else "Business B")
print("More consistent: ", "Business A" if sd_A < sd_B else "Business B")
```

### Level 4 - Coefficient of Variation(CV)

```python
# Calculate Coefficient of Variation
company_A_mean = 50000
company_A_sd = 5000

company_B_mean = 100000
company_B_sd = 8000

cv_A = (company_A_sd / company_A_mean) * 100
cv_B = (company_B_sd / company_B_mean) * 100

print("Company A CV:", cv_A, "%")
print("Company B CV:", cv_B, "%")

print(
    "Higher relative variability:",
    "Company A" if cv_A > cv_B else "Company B")
```

```python
# Business Interpretation
product_A_mean = 80000
product_A_sd = 8000

product_B_mean = 150000
product_B_sd = 18000
# Calculate CV
cv_A = (product_A_sd / product_A_mean) * 100
cv_B = (product_B_sd / product_B_mean) * 100
# Display statistics
print("Product A")
print("Mean Sales:", product_A_mean)
print("SD:", product_A_sd)
print("CV:", cv_A, "%")

print("\nProduct B")
print("Mean Sales:", product_B_mean)
print("SD:", product_B_sd)
print("CV:", cv_B, "%")
# Business comparison
print("\nBusiness Comparison")
print(
    "Higher average sales:",
    "Product A" if product_A_mean > product_B_mean else "Product B")
print(
    "Higher absolute variability:",
    "Product A" if product_A_sd > product_B_sd else "Product B")
print(
    "Higher relative variability:",
    "Product A" if cv_A > cv_B else "Product B")
```

### Level 5 - Outliers & IQR

```python
# Outlier Detection
data = [10, 12, 14, 15, 16, 18, 20, 22, 25, 80]
# Calculate quartiles
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
# Calculate IQR
IQR = Q3 - Q1
# Calculate bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# Identify outliers
outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Data:", data)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

```python
# Business Outlier
transactions = [500, 520, 510, 530, 515, 525, 540, 550, 5000]
Q1 = np.percentile(transactions, 25)
Q3 = np.percentile(transactions, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [
    x for x in transactions
    if x < lower_bound or x > upper_bound]
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Potential Outliers:", outliers)
```

### Level 6 - Z-Scores

```python
# Calculate Z-score
mean = 2000
sd = 400
x = 3200
z = (x - mean) / sd
print("Mean:", mean)
print("Standard Deviation:", sd)
print("Transaction:", x)
print("Z-score:", z)
```

```python
# Interpret Z-score
z = -2.5
print("Z-score:", z)
if z < 0:
    print("The observation is below the mean.")
print("The observation is 2.5 standard deviations below the mean.")
```

```python
# Compare Z-scores
# Product A
mean_A = 100
sd_A = 10
sales_A = 130
z_A = (sales_A - mean_A) / sd_A
# Product B
mean_B = 500
sd_B = 50
sales_B = 600
z_B = (sales_B - mean_B) / sd_B
print("Product A Z-score:", z_A)
print("Product B Z-score:", z_B)
print(
    "More unusual:",
    "Product A" if abs(z_A) > abs(z_B) else "Product B")
```

### Level 7 - Distribution & Skewness

```python
# Distribution Shape
A = [10, 11, 12, 12, 13, 14, 15]
B = [2, 3, 3, 4, 5, 6, 30]
C = [5, 20, 21, 22, 23, 24, 25]
datasets = {
    "A": A,
    "B": B,
    "C": C}
for name, data in datasets.items():
    mean = np.mean(data)
    median = np.median(data)

    print(name)
    print("Mean:", mean)
    print("Median:", median)
    print()
for name, data in datasets.items():
    mean = np.mean(data)
    median = np.median(data)
    if mean > median:
        shape = "Positively skewed"
    elif mean < median:
        shape = "Negatively skewed"
    else:
        shape = "Approximately symmetric"
    print(name, ":", shape)
```

```python
# Employee Salary Distribution
salaries = [25, 27, 28, 30, 32, 35, 200]

mean_salary = np.mean(salaries)
median_salary = np.median(salaries)

print("Salaries (₹000):", salaries)
print("Mean Salary: ₹", mean_salary, "k")
print("Median Salary: ₹", median_salary, "k")

if mean_salary > median_salary:
    print("Distribution : Positively skewed")
    print("Higher measure : Mean")
else:
    print("Distribution : Negatively skewed")
print("Better measure of typical salary : Median")
```

### Level 8 - Normal Distribution

```python
# Normal Distribution + Empirical Rule
mean = 100
sd = 15

range_1sd = (mean - sd, mean + sd)
range_2sd = (mean - 2 * sd, mean + 2 * sd)
range_3sd = (mean - 3 * sd, mean + 3 * sd)

print("Mean:", mean)
print("SD:", sd)

print("\nWithin ±1 SD:")
print("Range:", range_1sd)
print("Percentage: 68%")

print("\nWithin ±2 SD:")
print("Range:", range_2sd)
print("Percentage: 95%")

print("\nWithin ±3 SD:")
print("Range:", range_3sd)
print("Percentage: 99.7%")
```

```python
# Z-score for a Student's Score
mean = 70
sd = 10
score = 90
z = (score - mean) / sd

print("Mean:", mean)
print("SD:", sd)
print("Student Score:", score)
print("Z-score:", z)
print("\nThe student's score is", z, "standard deviations above the mean.")
```

### Level 10 - Business Statistics

```python
# Calculate: Mean, Median, Range, Variance, Standard Deviation
unit_A = [100, 110, 105, 95, 90]
unit_B = [100, 130, 70, 150, 50]
# Unit A Statistics
mean_A = np.mean(unit_A)
median_A = np.median(unit_A)
range_A = np.max(unit_A) - np.min(unit_A)
variance_A = np.var(unit_A)
sd_A = np.std(unit_A)

print("Unit A Statistics")
print("------------------")
print("Mean:", mean_A)
print("Median:", median_A)
print("Range:", range_A)
print("Variance:", variance_A)
print("Standard Deviation:", sd_A)
```

```python
# Unit B
mean_B = np.mean(unit_B)
median_B = np.median(unit_B)
range_B = np.max(unit_B) - np.min(unit_B)
variance_B = np.var(unit_B)
sd_B = np.std(unit_B)

print("Unit B Statistics")
print("------------------")
print("Mean:", mean_B)
print("Median:", median_B)
print("Range:", range_B)
print("Variance:", variance_B)
print("Standard Deviation:", sd_B)
```

```python
# Compare Unit A and Unit B
print("Business Comparison")
print("====================")
if mean_A > mean_B:
    result = "Unit A"
elif mean_B > mean_A:
    result = "Unit B"
else:
    result = "Both are equal"
print("Higher average performance:", result)
print(
    "Greater volatility:",
    "Unit A" if sd_A > sd_B else "Unit B")
print(
    "More consistent sales:",
    "Unit A" if sd_A < sd_B else "Unit B")
print(
    "Greater business risk based purely on sales variability:",
    "Unit A" if sd_A > sd_B else "Unit B")
```

```python
# Coefficient of Variation
cv_A = (sd_A / mean_A) * 100
cv_B = (sd_B / mean_B) * 100

print("Unit A CV:", cv_A, "%")
print("Unit B CV:", cv_B, "%")

print(
    "Greater relative volatility:",
    "Unit A" if cv_A > cv_B else "Unit B")
```

```python
# Final Business Interpretation
product_A = {
    "mean_sales": 100000,
    "median_sales": 98000,
    "sd": 8000,
    "cv": 8,
    "outliers": 0,
    "z_score": 1.2}
product_B = {
    "mean_sales": 150000,
    "median_sales": 120000,
    "sd": 40000,
    "cv": 26.67,
    "outliers": 2,
    "z_score": 3.8}
print("Product A")
print("Mean Sales:", product_A["mean_sales"])
print("Median Sales:", product_A["median_sales"])
print("SD:", product_A["sd"])
print("CV:", product_A["cv"], "%")
print("Outliers:", product_A["outliers"])
print("Z-score:", product_A["z_score"])

print("\nProduct B")
print("Mean Sales:", product_B["mean_sales"])
print("Median Sales:", product_B["median_sales"])
print("SD:", product_B["sd"])
print("CV:", product_B["cv"], "%")
print("Outliers:", product_B["outliers"])
print("Z-score:", product_B["z_score"])
```

```python
# Business Comparison
better_performance = (
    "Product A"
    if product_A["mean_sales"] > product_B["mean_sales"]
    else "Product B")
more_stable = (
    "Product A"
    if product_A["cv"] < product_B["cv"]
    else "Product B")
greater_risk = (
    "Product A"
    if (
        product_A["cv"] > product_B["cv"]
        and product_A["outliers"] > product_B["outliers"])
    else "Product B")
print("Better average performance:", better_performance)
print("More stable:", more_stable)
print("Greater sales risk:", greater_risk)
```

```python
# Final Business Conclusion
print("""Product B performs better in terms of average sales because its mean sales are ₹150000 compared with ₹100000 for Product A.

However, Product A is much more stable. Its standard deviation is ₹8000 and its CV is only 8%, compared with ₹40000 SD and 26.67% CV
for Product B.

Product B also has 2 outliers compared with none for Product A and its unusual sale has a Z-score of 3.8 compared with 1.2 for Product A.
This indicates that Product B has more extreme and unusual sales values.

Therefore, Product B has higher sales performance but also carries  greater sales risk, while Product A has lower average sales 
but is considerably more stable and predictable.""")
```
