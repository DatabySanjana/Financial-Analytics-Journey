```python
# Demand: The quantity of a good consumers are willing and able to buy at different prices, ceteris paribus.
#Price increases from ₹100 to ₹120 and quantity demanded falls from 1000 to 800 units.
old_price = 100
new_price = 120
old_quantity = 1000
new_quantity = 800
quantity_change = new_quantity - old_quantity
percentage_change = (quantity_change / old_quantity) * 100
print("Change in quantity:", quantity_change)
print("Percentage change:", percentage_change, "%")
```

```python
# Supply: The quantity of a good producers are willing and able to sell at different prices, ceteris paribus.
# Price rises from ₹50 to ₹70 and producers increase supply from 500 to 800 units.
old_price = 50
new_price = 70
old_supply = 500
new_supply = 800
supply_change = new_supply - old_supply
percentage_change = (supply_change / old_supply) * 100
print("Change in supply:", supply_change)
print("Percentage change:", percentage_change, "%")
```

```python
# Market equilibrium: The point where quantity demanded equals quantity supplied.
prices = [20, 30, 40, 50]
demand = [1000, 900, 800, 650]
supply = [400, 600, 800, 1000]
for i in range(len(prices)):
    if demand[i] == supply[i]:
        print("Equilibrium Price:", prices[i])
        print("Equilibrium Quantity:", demand[i])
```

```python
# Price Elasticity of Demand: Measures how responsive quantity demanded is to a change in price.
# Price increases by 10% and quantity demanded decreases by 25%.
percentage_change_quantity = -25
percentage_change_price = 10
ped = percentage_change_quantity / percentage_change_price
print("Price Elasticity of Demand:", ped)
print("Absolute Elasticity:", abs(ped))
```

```python
# Income Elasticity of Demand: Measures how demand changes when consumer income changes.
# Income increases by 20% and demand for restaurant meals increases by 10%.
percentage_change_demand = 10
percentage_change_income = 20
yed = percentage_change_demand / percentage_change_income
print("Income Elasticity of Demand:", yed)
```

```python
# Substitutes: Goods that can be used in place of each other.
# Coffee and tea are substitutes. Coffee price increases by 15%. What happens to demand for tea?
coffee_price_change = 15
if coffee_price_change > 0:
    print("Demand for tea is likely to increase.")
```

```python
# Complements: Goods that are consumed or used together.
# The price of smartphones rises significantly. What could happen to demand for smartphone accessories?
smartphone_price_change = 20
if smartphone_price_change > 0:
    print("Demand for smartphone accessories is likely to decrease.")
```

```python
# Marginal Cost: The additional cost of producing one additional unit.
old_cost = 50000
new_cost = 54000
old_quantity = 100
new_quantity = 110
marginal_cost = (new_cost - old_cost) / (new_quantity - old_quantity)
print("Marginal Cost per unit:", marginal_cost)
```

```python
# Marginal Revenue: The additional revenue generated from selling one additional unit.
old_revenue = 20000
new_revenue = 21500
old_quantity = 100
new_quantity = 110
marginal_revenue = (new_revenue - old_revenue) / (new_quantity - old_quantity)
print("Marginal Revenue per unit:", marginal_revenue)
```

```python
# Opportunity Cost: The value of the next-best alternative given up when making a decision.
returns = {
    "Project A": 80000,
    "Project B": 120000,
    "Project C": 60000}
chosen_project = "Project A"
alternative_returns = [
    value for project, value in returns.items()
    if project != chosen_project]
opportunity_cost = max(alternative_returns)
print("Opportunity Cost:", opportunity_cost)
```

```python
# Demand, Supply & Equilibrium
prices = [100, 120, 140, 160]
quantity_demanded = [10000, 8000, 6000, 4000]
quantity_supplied = [4000, 6000, 6000, 8000]
for i in range(len(prices)):
    if quantity_demanded[i] == quantity_supplied[i]:
        equilibrium_price = prices[i]
        equilibrium_quantity = quantity_demanded[i]
shortage_at_120 = quantity_demanded[1] - quantity_supplied[1]
surplus_at_160 = quantity_supplied[3] - quantity_demanded[3]
percentage_change_demand = (
    (quantity_demanded[2] - quantity_demanded[0])
    / quantity_demanded[0]
) * 100
print("Equilibrium Price:", equilibrium_price)
print("Equilibrium Quantity:", equilibrium_quantity)
print("Shortage at ₹120:", shortage_at_120)
print("Surplus at ₹160:", surplus_at_160)
print("Change in demand:", percentage_change_demand, "%")
```

```python
# Price Elasticity + Revenue
# Calculate: Equilibrium price, Equilibrium quantity, Shortage at ₹120, Surplus at ₹160, Percentage decrease in quantity demanded from ₹100 to ₹140
price = 2000
quantity = 5000
price_change = 10
elasticity = 1.5
quantity_change = -elasticity * price_change
new_quantity = quantity * (1 + quantity_change / 100)
new_price = price * (1 + price_change / 100)
current_revenue = price * quantity
new_revenue = new_price * new_quantity
print("Quantity change:", quantity_change, "%")
print("New quantity:", new_quantity)
print("New price:", new_price)
print("Current revenue:", current_revenue)
print("New revenue:", new_revenue)
```

```python
# Marginal Cost + Marginal Revenue + Profit Decision
# Calculate: Marginal cost, Marginal revenue, Marginal profit, Should the company consider producing the additional units based on this comparison?
old_quantity = 1000
new_quantity = 1100

old_cost = 500000
new_cost = 560000

old_revenue = 800000
new_revenue = 880000

marginal_cost = (new_cost - old_cost) / (new_quantity - old_quantity)
marginal_revenue = (new_revenue - old_revenue) / (new_quantity - old_quantity)
marginal_profit = marginal_revenue - marginal_cost
print("Marginal Cost:", marginal_cost)
print("Marginal Revenue:", marginal_revenue)
print("Marginal Profit:", marginal_profit)
print("Since MR > MC, producing additional units is financially attractive, assuming other conditions remain unchanged.")
```

```python
# Opportunity Cost + Economies of Scale
# Calculate: Opportunity cost, Average cost at each output, Is there evidence of economies of scale?
returns = {
    "Project A": 800000,
    "Project B": 1200000,
    "Project C": 1000000}
chosen_project = "Project C"
opportunity_cost = max(
    value for project, value in returns.items()
    if project != chosen_project)
output = [1000, 2000, 4000]
total_cost = [200000, 340000, 600000]
print("Opportunity Cost:", opportunity_cost)
for i in range(len(output)):
    average_cost = total_cost[i] / output[i]
    print("Output:", output[i], "| Average Cost:", average_cost)
```

```python
# Inflation + Profit Margin
# Calculate: Current total cost, Current profit, New total cost, New profit, Current profit margin, New profit margin
revenue = 10000000
raw_material = 4000000
labour = 2000000
transportation = 1000000
other_costs = 1000000
new_raw_material = raw_material * 1.15
new_labour = labour * 1.08
new_transportation = transportation * 1.12
new_other_costs = other_costs * 1.05
old_total_cost = raw_material + labour + transportation + other_costs
new_total_cost = (
    new_raw_material
    + new_labour
    + new_transportation
    + new_other_costs)
old_profit = revenue - old_total_cost
new_profit = revenue - new_total_cost

old_margin = (old_profit / revenue) * 100
new_margin = (new_profit / revenue) * 100

print("Current Total Cost:", old_total_cost)
print("Current Profit:", old_profit)
print("New Total Cost:", new_total_cost)
print("New Profit:", new_profit)
print("Current Profit Margin:", old_margin, "%")
print("New Profit Margin:", new_margin, "%")
```

```python
# Interest Rates + Loan Cost + Investment
# Calculate: Old annual interest, New annual interest, Increase in annual interest expense, 
#            If expected annual project profit before interest is ₹50 lakh, calculate profit after interest under both scenarios.
loan = 50000000
old_rate = 7
new_rate = 9

old_interest = loan * old_rate / 100
new_interest = loan * new_rate / 100

project_profit_before_interest = 5000000

old_profit_after_interest = project_profit_before_interest - old_interest
new_profit_after_interest = project_profit_before_interest - new_interest

print("Old Annual Interest:", old_interest)
print("New Annual Interest:", new_interest)
print("Increase in Interest Expense:", new_interest - old_interest)

print("Profit after old interest:", old_profit_after_interest)
print("Profit after new interest:", new_profit_after_interest)
```

```python
# Exchange Rate + Import Cost + Pricing
# Calculate: Old import cost, New import cost, Increase in import cost, Percentage increase in import cost
#            If the company cannot increase its selling price, what happens to profit margin?
usd_imports = 500000

old_rate = 82
new_rate = 88

old_import_cost = usd_imports * old_rate
new_import_cost = usd_imports * new_rate

increase = new_import_cost - old_import_cost
percentage_increase = (increase / old_import_cost) * 100

print("Old Import Cost:", old_import_cost)
print("New Import Cost:", new_import_cost)
print("Increase in Import Cost:", increase)
print("Percentage Increase:", percentage_increase, "%")
print("Impact: Profit margin decreases if selling price remains unchanged.")
```

```python
# GDP + Unemployment + Consumer Demand
# Calculate: Current revenue, New average spending, New revenue, Revenue loss, Percentage revenue decline
customers = 100000
average_spending = 5000

spending_decline = 12

current_revenue = customers * average_spending

new_average_spending = average_spending * (1 - spending_decline / 100)
new_revenue = customers * new_average_spending

revenue_loss = current_revenue - new_revenue
percentage_decline = (revenue_loss / current_revenue) * 100

print("Current Revenue:", current_revenue)
print("New Average Spending:", new_average_spending)
print("New Revenue:", new_revenue)
print("Revenue Loss:", revenue_loss)
print("Revenue Decline:", percentage_decline, "%")
```

```python
# Macroeconomic business analysis: Evaluates how GDP, inflation, interest rates, exchange rates, 
#                                  unemployment and government/central-bank policies affect business performance.
# Calculate: Current profit, New imported component cost, New other costs, New interest expense, New revenue, New total cost, New profit, 
#            Current vs new profit margin, Profit decline, Give 3 management recommendations.
# Current figures
current_revenue = 100000000
imported_components = 40000000
other_costs = 40000000
loan = 20000000
old_interest_rate = 8
new_interest_rate = 10
# Current interest
old_interest = loan * old_interest_rate / 100
# Current total cost and profit
current_total_cost = imported_components + other_costs + old_interest
current_profit = current_revenue - current_total_cost
# Economic changes
revenue_decline = 10
import_cost_increase = 8
other_cost_increase = 10
new_revenue = current_revenue * (1 - revenue_decline / 100)
new_imported_components = imported_components * (
    1 + import_cost_increase / 100)
new_other_costs = other_costs * (
    1 + other_cost_increase / 100)
new_interest = loan * new_interest_rate / 100
# New total cost and profit
new_total_cost = (
    new_imported_components
    + new_other_costs
    + new_interest)
new_profit = new_revenue - new_total_cost
# Margins
current_margin = (current_profit / current_revenue) * 100
new_margin = (new_profit / new_revenue) * 100
profit_decline = current_profit - new_profit

print("Current Profit:", current_profit)
print("New Imported Component Cost:", new_imported_components)
print("New Other Costs:", new_other_costs)
print("New Interest Expense:", new_interest)
print("New Revenue:", new_revenue)
print("New Total Cost:", new_total_cost)
print("New Profit:", new_profit)
print("Current Profit Margin:", current_margin, "%")
print("New Profit Margin:", new_margin, "%")
print("Profit Decline:", profit_decline)
```

```python
# Q12 — Deflation
# Question: A retail company sells 10000 units at an average price of ₹2000.
# Due to deflation: Average selling price falls by 8% & Quantity sold falls by 5%
# Calculate: Current revenue, New selling price, New quantity sold, New revenue, Percentage change in revenue
price = 2000
quantity = 10000
price_change = -8
quantity_change = -5
current_revenue = price * quantity
new_price = price * (1 + price_change / 100)
new_quantity = quantity * (1 + quantity_change / 100)
new_revenue = new_price * new_quantity
revenue_change = new_revenue - current_revenue
percentage_revenue_change = (
    revenue_change / current_revenue
) * 100
print("Current Revenue:", current_revenue)
print("New Selling Price:", new_price)
print("New Quantity:", new_quantity)
print("New Revenue:", new_revenue)
print("Revenue Change:", revenue_change)
print("Percentage Revenue Change:", percentage_revenue_change, "%")
print("Business interpretation:Deflation can appear beneficial because prices fall, but persistent deflation can cause consumers to delay purchases, reducing business revenue and economic activity.")
```

```python
# Fiscal Policy
# Question: Government launches an infrastructure program worth ₹500 crore.
# 60% goes to construction businesses, 20% goes to equipment suppliers, 20% goes to wages.
# Calculate: Construction spending, Equipment spending, Wage spending, Explain likely impact on GDP and employment.
government_spending = 500  # crore
construction_share = 60
equipment_share = 20
wage_share = 20
construction_spending = government_spending * construction_share / 100
equipment_spending = government_spending * equipment_share / 100
wage_spending = government_spending * wage_share / 100
print("Construction Spending:", construction_spending, "crore")
print("Equipment Spending:", equipment_spending, "crore")
print("Wage Spending:", wage_spending, "crore")
print("\nLikely impact:")
print("Government spending → Business activity ↑ → Employment ↑")
print("Higher economic activity → GDP growth may increase")
```

```python
# Monetary Policy
# Question: Central bank raises policy interest rate from 6% to 7.5%. A company has a ₹10 crore floating-rate loan. 
# Assume the loan rate increases by the same 1.5 percentage points.
# Calculate: Old annual interest expense, New annual interest expense, Increase in interest expense, Explain the monetary policy transmission mechanism.
loan = 100000000
old_rate = 6
new_rate = 7.5
old_interest = loan * old_rate / 100
new_interest = loan * new_rate / 100
increase = new_interest - old_interest
print("Old Interest Expense:", old_interest)
print("New Interest Expense:", new_interest)
print("Increase in Interest Expense:", increase)
print("\nTransmission mechanism:")
print("Interest rates ↑ → Borrowing ↓ → Spending ↓")
print("→ Aggregate demand ↓ → Inflationary pressure ↓")
```
