```python
# Accounting Equation
# Q1. A company has: Cash = ₹20 Cr, Inventory = ₹15 Cr, Equipment = ₹25 Cr, Bank Loan = ₹30 Cr
# Calculate Total Assets, Total Liabilities and Equity and verify the accounting equation.
cash = 12
accounts_receivable = 8
inventory = 10
property_equipment = 30

accounts_payable = 7
long_term_debt = 18
# Calculate Total Assets
total_assets = cash + accounts_receivable + inventory + property_equipment
# Calculate Total Liabilities
total_liabilities = accounts_payable + long_term_debt
# Calculate Equity
equity = total_assets - total_liabilities
print("Total Assets:", total_assets, "Cr")
print("Total Liabilities:", total_liabilities, "Cr")
print("Equity:", equity, "Cr")
# Verify Accounting Equation
print("Assets =", total_assets)
print("Liabilities + Equity =", total_liabilities + equity)
print("Accounting Equation Balanced:", total_assets == total_liabilities + equity)
```

```python
# Section 2 - Financial Statements
# Income Statement
# Q4. Given: Revenue = ₹500 Cr, COGS = ₹300 Cr, Operating Expenses = ₹100 Cr, Interest = ₹20 Cr, Tax = ₹20 Cr
# Calculate: Gross Profit -> Operating Profit -> Profit Before Tax -> Net Profit
revenue = 500
cogs = 300
salaries = 50
marketing = 20
rent = 15
depreciation = 10
interest = 12
tax = 20

gross_profit = revenue - cogs
operating_expenses = (salaries + marketing + rent + depreciation)
operating_profit = gross_profit - operating_expenses
profit_before_tax = operating_profit - interest
net_profit = profit_before_tax - tax

print("Gross Profit:", gross_profit, "Cr")
print("Operating Expenses:", operating_expenses, "Cr")
print("Operating Profit / EBIT:", operating_profit, "Cr")
print("Profit Before Tax:", profit_before_tax, "Cr")
print("Net Profit:", net_profit, "Cr")
```

```python
# Balance Sheet
# Q7. Create a simplified Balance Sheet:
# Calculate: Current Assets, Non-current Assets, Total Assets, Current Liabilities, Non-current Liabilities, Total Liabilities, Total Equity
# Verify: Assets = Liabilities + Equity
cash = 40
accounts_receivable = 60
inventory = 80
property_equipment = 220

accounts_payable = 70
short_term_debt = 30
long_term_debt = 100

share_capital = 100
retained_earnings = 100

current_assets = cash + accounts_receivable + inventory
non_current_assets = property_equipment

total_assets = current_assets + non_current_assets

current_liabilities = accounts_payable + short_term_debt
non_current_liabilities = long_term_debt

total_liabilities = current_liabilities + non_current_liabilities

total_equity = share_capital + retained_earnings

print("Current Assets:", current_assets)
print("Non-current Assets:", non_current_assets)
print("Total Assets:", total_assets)

print("Current Liabilities:", current_liabilities)
print("Non-current Liabilities:", non_current_liabilities)
print("Total Liabilities:", total_liabilities)
print("Total Equity:", total_equity)
print(
    "Balanced:",
    total_assets == total_liabilities + total_equity)
```

```python
# PROFITABILITY RATIOS
# Gross Margin
# Q11. Calculate: Gross Margin = Gross Profit / Revenue × 100
gross_margin_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Revenue": [400, 450, 500, 580],
    "Gross Profit": [160, 180, 190, 203]})
gross_margin_df["Gross Margin %"] = (gross_margin_df["Gross Profit"] / gross_margin_df["Revenue"] * 100)
print(gross_margin_df)
```

```python
# ROA & ROE
# Q14 — ROA
# ROA = Net Income / Average Assets × 100
roa_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Net Income": [40, 45, 48, 50],
    "Average Assets": [400, 450, 520, 600]})
roa_df["ROA %"] = (roa_df["Net Income"] / roa_df["Average Assets"] * 100)
print(roa_df)
print("\nInterpretation: Asset efficiency is declining.")
```

```python
# Q15 — ROE
roe_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Net Income": [40, 45, 48, 50],
    "Average Equity": [200, 220, 240, 260]})
roe_df["ROE %"] = (roe_df["Net Income"] / roe_df["Average Equity"] * 100)
print(roe_df)
print("Interpretation: Shareholder returns are slightly declining after 2023.")
```

```python
# LIQUIDITY
# Q16 — Current Ratio
current_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Current Assets": [150, 180, 210, 250],
    "Current Liabilities": [100, 120, 150, 200]})
current_df["Current Ratio"] = (current_df["Current Assets"] / current_df["Current Liabilities"])
print(current_df)
```

```python
plt.figure(figsize=(9, 5))
plt.plot(current_df["Year"], current_df["Current Ratio"], marker="o")
plt.title("Current Ratio Trend")
plt.ylabel("Current Ratio")
plt.show()
print("\nInterpretation: Short-term liquidity is weakening.")
```

```python
# Q17 — Quick Ratio
# Quick Ratio = (Current Assets − Inventory) / Current Liabilities
quick_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Current Assets": [150, 180, 210, 250],
    "Inventory": [50, 70, 90, 120],
    "Current Liabilities": [100, 120, 150, 200]})
quick_df["Quick Ratio"] = ((quick_df["Current Assets"] - quick_df["Inventory"]) / quick_df["Current Liabilities"])
print(quick_df)
```

```python
plt.figure(figsize=(9, 5))
plt.plot(current_df["Year"], current_df["Current Ratio"], marker="o", label="Current Ratio")
plt.plot(quick_df["Year"], quick_df["Quick Ratio"], marker="o", label="Quick Ratio")
plt.title("Current Ratio vs Quick Ratio")
plt.ylabel("Ratio")
plt.legend()
plt.show()
print("\nInterpretation: Quick ratio is lower because inventory is excluded. Inventory cannot generally be converted into cash as quickly as cash or receivables.")
```

```python
# LEVERAGE
# Q18. Debt-to-Equity
# D/E = Debt / Equity
de_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Debt": [150, 180, 230, 300],
    "Equity": [200, 220, 240, 260]})
de_df["Debt-to-Equity"] = (de_df["Debt"] / de_df["Equity"])
print(de_df)
print("\nInterpretation: Financial leverage is becoming more concerning because debt is growing faster than equity.")
```

```python
# Q19 — Debt Ratio
# Debt Ratio = Total Debt / Total Assets × 100
debt_ratio_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Debt": [150, 180, 230, 300],
    "Assets": [400, 450, 520, 600]})
debt_ratio_df["Debt Ratio %"] = (debt_ratio_df["Debt"] / debt_ratio_df["Assets"] * 100)
print(debt_ratio_df)
print("\nInterpretation: Increasing debt ratio means a larger proportion of assets is financed through debt, increasing financial risk.")
```

```python
# Q20 — Interest Coverage
# Interest Coverage = EBIT / Interest Expense
interest_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "EBIT": [60, 72, 70, 69],
    "Interest Expense": [10, 12, 18, 25]})
interest_df["Interest Coverage"] = (interest_df["EBIT"] / interest_df["Interest Expense"])
print(interest_df)
```

```python
plt.figure(figsize=(9, 5))
plt.plot(interest_df["Year"], interest_df["Interest Coverage"], marker="o")
plt.title("Interest Coverage Trend")
plt.ylabel("Times")
plt.show()
print("\nInterpretation: The company has become less capable of covering interest expenses from operating earnings, indicating increasing financial risk.")
```

```python
# EFFICIENCY
# Q21 — Asset Turnover
# Asset Turnover = Revenue / Average Assets
asset_turnover_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Revenue": [400, 450, 500, 580],
    "Average Assets": [400, 450, 520, 600]})
asset_turnover_df["Asset Turnover"] = (asset_turnover_df["Revenue"] / asset_turnover_df["Average Assets"])
print(asset_turnover_df)
print("\nInterpretation: Asset utilization has weakened slightly compared with the earlier years.")
```

```python
# Q22 — Inventory Turnover
# Inventory Turnover = COGS / Average Inventory
inventory_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "COGS": [240, 270, 300, 350],
    "Average Inventory": [40, 50, 65, 90]})
inventory_df["Inventory Turnover"] = (inventory_df["COGS"] / inventory_df["Average Inventory"])
print(inventory_df)
```

```python
plt.figure(figsize=(9, 5))
plt.plot(inventory_df["Year"], inventory_df["Inventory Turnover"], marker="o")
plt.title("Inventory Turnover Trend")
plt.ylabel("Times")
plt.show()
print("\nInterpretation: Declining turnover can indicate slower inventory movement, excess inventory, weak demand or inventory management problems.")
```

```python
# Q23 — Receivables Turnover
# Receivables Turnover = Revenue / Average Receivables
receivables_df = pd.DataFrame({
    "Year": [2022, 2023, 2024, 2025],
    "Revenue": [400, 450, 500, 580],
    "Average Receivables": [40, 50, 65, 85]})
receivables_df["Receivables Turnover"] = (receivables_df["Revenue"] / receivables_df["Average Receivables"])
print(receivables_df)
print("\nInterpretation: The decline suggests the company is collecting receivables less efficiently.")
```

```python
# Calculate Financial Ratios
# Extract values
revenue = company_df.loc[company_df["Metric"] == "Revenue", ["2024", "2025"]].iloc[0]

gross_profit = company_df.loc[company_df["Metric"] == "Gross Profit", ["2024", "2025"]].iloc[0]
operating_profit = company_df.loc[company_df["Metric"] == "Operating Profit", ["2024", "2025"]].iloc[0]
net_profit = company_df.loc[company_df["Metric"] == "Net Profit", ["2024", "2025"]].iloc[0]

current_assets = company_df.loc[company_df["Metric"] == "Current Assets", ["2024", "2025"]].iloc[0]
current_liabilities = company_df.loc[company_df["Metric"] == "Current Liabilities", ["2024", "2025"]].iloc[0]

debt = company_df.loc[company_df["Metric"] == "Debt", ["2024", "2025"]].iloc[0]
equity = company_df.loc[company_df["Metric"] == "Equity", ["2024", "2025"]].iloc[0]
assets = company_df.loc[company_df["Metric"] == "Assets", ["2024", "2025"]].iloc[0]
ebit = company_df.loc[company_df["Metric"] == "EBIT", ["2024", "2025"]].iloc[0]
interest = company_df.loc[company_df["Metric"] == "Interest", ["2024", "2025"]].iloc[0]
cfo = company_df.loc[company_df["Metric"] == "CFO",["2024", "2025"]].iloc[0]
# Calculate Ratios
gross_margin = gross_profit / revenue * 100
operating_margin = operating_profit / revenue * 100
net_margin = net_profit / revenue * 100
current_ratio = current_assets / current_liabilities
debt_to_equity = debt / equity
debt_ratio = debt / assets * 100
interest_coverage = ebit / interest
roa = net_profit / assets * 100
roe = net_profit / equity * 100
# Create Ratio Table
ratio_df = pd.DataFrame({
    "Ratio": ["Gross Margin", "Operating Margin", "Net Margin", "Current Ratio", "Debt-to-Equity", "Debt Ratio", "Interest Coverage", "ROA", "ROE"],
    "2024": [
        gross_margin["2024"],
        operating_margin["2024"],
        net_margin["2024"],
        current_ratio["2024"],
        debt_to_equity["2024"],
        debt_ratio["2024"],
        interest_coverage["2024"],
        roa["2024"],
        roe["2024"]],
    "2025": [
        gross_margin["2025"],
        operating_margin["2025"],
        net_margin["2025"],
        current_ratio["2025"],
        debt_to_equity["2025"],
        debt_ratio["2025"],
        interest_coverage["2025"],
        roa["2025"],
        roe["2025"]]})
ratio_df
```

```python
# Formatted Ratio Output
print("FINANCIAL RATIO COMPARISON")
for _, row in ratio_df.iterrows():
    print(
        f"{row['Ratio']:<22} "
        f"2024: {row['2024']:.2f}    "
        f"2025: {row['2025']:.2f}")
```
