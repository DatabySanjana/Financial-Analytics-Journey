##### 1. Percentages
Percentage tells us how much one value represents out of a total.
 - Percentages can be used to understand: Proportion of sales, Market share, Customer distribution, Expense contribution, Business performance

```python
# Q1 — Percentage
# A company has total sales of ₹800,000. Product A generated ₹200,000. Calculate Product A's percentage of total sales.
total_sales = 800000
product_a_sales = 200000
percentage = (product_a_sales / total_sales) * 100
print(percentage)
```

```python
# Q2 — Percentage
# A company has 5,000 customers. 1,250 customers purchased Product A. What percentage of customers purchased Product A?
total_customers = 5000
product_a_customers = 1250
percentage = (product_a_customers / total_customers) * 100
print(percentage)
```

```python
# Q3 — Percentage Change
# Revenue increased from ₹600,000 to ₹750,000. Calculate the percentage change.
old_revenue = 600000
new_revenue = 750000
percentage_change = ((new_revenue - old_revenue) / old_revenue) * 100
print(percentage_change)
```

```python
# Q4 — Percentage Change
# Sales decreased from ₹900,000 to ₹720,000. Calculate the percentage change.
old_sales = 900000
new_sales = 720000
percentage_change = ((new_sales - old_sales) / old_sales) * 100
print(percentage_change)
```

```python
# Q5 — Ratio
# A business has: Online Sales = ₹300,000 & Offline Sales = ₹500,000. Find the Online : Offline ratio in simplest form.
online_sales = 300000
offline_sales = 500000
ratio = online_sales / offline_sales
print(ratio)
```

```python
# Q6 — Proportion
# If 5 employees generate ₹500,000 in sales, how much sales would 8 employees generate, if productivity remains constant?
employees_1 = 5
sales_1 = 500000
employees_2 = 8
sales_per_employee = sales_1 / employees_1
sales_2 = sales_per_employee * employees_2
print(sales_2)
```

```python
# Q7 — Average
# Monthly sales are: ₹100000, ₹150000, ₹200000, ₹250000. Calculate average monthly sales.
monthly_sales = [100000, 150000, 200000, 250000]
average_sales = sum(monthly_sales) / len(monthly_sales)
print(average_sales)
```

```python
# Q8 — Weighted Average
# Product A: Price = ₹100, Units = 200 & Product B: Price = ₹200, Units = 800. Calculate the weighted average selling price.
price_a = 100
units_a = 200
price_b = 200
units_b = 800
weighted_average = (
    (price_a * units_a) + (price_b * units_b)
) / (units_a + units_b)
print(weighted_average)
```

```python
# Q9 — Growth Rate
# A company's customer base increased from 10,000 to 12,500. Calculate customer growth rate.
total_expenses = 500000
marketing_expenses = 75000
marketing_percentage = (marketing_expenses / total_expenses) * 100
print(marketing_percentage)
```

```python
# Q10 — CAGR
# A company's revenue increased from ₹10 lakh to ₹20 lakh over 3 years. Calculate the CAGR.
beginning_revenue = 10   # lakh
ending_revenue = 20      # lakh
years = 3
cagr = ((ending_revenue / beginning_revenue) ** (1 / years) - 1) * 100
print(cagr)
```

```python
# Q11 — Percentage
# A company's total expenses are ₹5 lakh. Marketing expenses are ₹75,000. What percentage of total expenses is marketing?
total_expenses = 500000
marketing_expenses = 75000
marketing_percentage = (marketing_expenses / total_expenses) * 100
print(marketing_percentage)
```

```python
# Q12 — Average
# A company's quarterly sales are: Q1 = ₹10 lakh, Q2 = ₹12 lakh, Q3 = ₹15 lakh, Q4 = ₹13 lakh. Calculate average quarterly sales.
quarterly_sales = [10, 12, 15, 13]  # lakh
average_sales = sum(quarterly_sales) / len(quarterly_sales)
print(average_sales)
```

### Business-Based Practice

```python
# Q1 — Revenue Growth
#A company's revenue was: 2024 → ₹40 lakh; 2025 → ₹50 lakh; 2026 → ₹62.5 lakh.
#Calculate: 2025 growth rate, 2026 growth rate, Which year had higher growth?
revenue_2024 = 40   # lakh
revenue_2025 = 50
revenue_2026 = 62.5
growth_2025 = ((revenue_2025 - revenue_2024) / revenue_2024) * 100
growth_2026 = ((revenue_2026 - revenue_2025) / revenue_2025) * 100
print("2025 Growth:", growth_2025, "%")
print("2026 Growth:", growth_2026, "%")
if growth_2025 > growth_2026:
    print("2025 had higher growth")
elif growth_2026 > growth_2025:
    print("2026 had higher growth")
else:
    print("Both years had the same growth")
```

```python
# Q2 — Customer Growth
# A company had 20,000 customers last year and 24,000 customers this year. Calculate customer growth.
old_customers = 20000
new_customers = 24000
growth = ((new_customers - old_customers) / old_customers) * 100
print("Customer Growth:", growth, "%")
```

```python
# Q3 — Weighted Average
# A company sells two products: Product A: Selling Price = ₹500, Units Sold = 1,000 & Product B: Selling Price = ₹800, Units Sold = 4,000
#Calculate: Simple average selling price, Weighted average selling price.
price_a = 500
units_a = 1000
price_b = 800
units_b = 4000
simple_average = (price_a + price_b) / 2
weighted_average = (
    (price_a * units_a) + (price_b * units_b)
) / (units_a + units_b)
print("Simple Average:", simple_average)
print("Weighted Average:", weighted_average)
```

```python
# Q4 — Business Ratio
# A company has: Online Sales = ₹30 lakh & Retail Sales = ₹45 lakh
# Calculate: Online : Retail ratio, Online sales as a percentage of total sales, Retail sales as a percentage of total sales
online_sales = 30   # lakh
retail_sales = 45   # lakh
total_sales = online_sales + retail_sales
online_retail_ratio = f"{online_sales / 15:.0f}:{retail_sales / 15:.0f}"
online_percentage = (online_sales / total_sales) * 100
retail_percentage = (retail_sales / total_sales) * 100
print("Online : Retail =", online_retail_ratio)
print("Online %:", online_percentage)
print("Retail %:", retail_percentage)
```

```python
# Q5 — CAGR
# A company's revenue was ₹25 lakh in 2023 and ₹50 lakh in 2026. Calculate the CAGR.
beginning_revenue = 25   # lakh
ending_revenue = 50      # lakh
years = 3
cagr = ((ending_revenue / beginning_revenue) ** (1 / years) - 1) * 100
print("CAGR:", cagr, "%")
```

```python
# Q6 — Management Interpretation
# Company A: Revenue growth = 15% & Company B: Revenue growth = 8%. Which company has stronger revenue growth?
company_a_growth = 15
company_b_growth = 8
if company_a_growth > company_b_growth:
    print("Company A has stronger revenue growth.")
else:
    print("Company B has stronger revenue growth.")
```

```python
# Q7 — Growth vs Decline
# A company's customer base changed: 2025 = 50000 & 2026 = 45000. Calculate the percentage change. Is this growth or decline?
customers_2025 = 50000
customers_2026 = 45000
percentage_change = (
    (customers_2026 - customers_2025)
    / customers_2025
) * 100
print("Percentage Change:", percentage_change, "%")
if percentage_change > 0:
    print("This is growth.")
else:
    print("This is a decline.")
```

### Q8 — Combined Business Analysis
A company reports:

2025 Revenue = ₹80 lakh
2026 Revenue = ₹100 lakh

2025 Customers = 10,000
2026 Customers = 12,500

Calculate:
1. Revenue growth
2. Customer growth
3. Revenue per customer for both years
4. Percentage change in revenue per customer


```python
revenue_2025 = 80       # lakh
revenue_2026 = 100
customers_2025 = 10000
customers_2026 = 12500
# 1. Revenue Growth
revenue_growth = (
    (revenue_2026 - revenue_2025)
    / revenue_2025
) * 100
# 2. Customer Growth
customer_growth = (
    (customers_2026 - customers_2025)
    / customers_2025
) * 100
# 3. Revenue per Customer
revenue_per_customer_2025 = (
    revenue_2025 * 100000
) / customers_2025
revenue_per_customer_2026 = (
    revenue_2026 * 100000
) / customers_2026
# 4. Change in Revenue per Customer
revenue_per_customer_change = (
    (revenue_per_customer_2026 - revenue_per_customer_2025)
    / revenue_per_customer_2025
) * 100
print("Revenue Growth:", revenue_growth, "%")
print("Customer Growth:", customer_growth, "%")
print("2025 Revenue/Customer:", revenue_per_customer_2025)
print("2026 Revenue/Customer:", revenue_per_customer_2026)
print("Revenue/Customer Change:", revenue_per_customer_change, "%")
```

Goal: Sales → Revenue → Costs → Profit → Margins → Break-even → Business Insights

```python
# Q1 — Revenue + Costs
# A company sells 2,000 units at ₹500 each. Fixed Cost = ₹200000 & Variable Cost per Unit = ₹150
# Calculate: Revenue, Variable Cost, Total Cost, Profit
units_sold = 2000
selling_price = 500
fixed_cost = 200000
variable_cost_per_unit = 150
# 1. Revenue
revenue = units_sold * selling_price
# 2. Variable Cost
variable_cost = units_sold * variable_cost_per_unit
# 3. Total Cost
total_cost = fixed_cost + variable_cost
# 4. Profit
profit = revenue - total_cost
print("Revenue:", revenue)
print("Variable Cost:", variable_cost)
print("Total Cost:", total_cost)
print("Profit:", profit)
```

```python
# Q2 — Gross Profit + Gross Margin
# A company has: Revenue = ₹1000000 & COGS = ₹6,00,000
# Calculate: Gross Profit, Gross Profit Margin. Interpret: What does the gross margin tell management?
revenue = 1000000
cogs = 600000
# Gross Profit
gross_profit = revenue - cogs
# Gross Profit Margin
gross_profit_margin = (gross_profit / revenue) * 100
print("Gross Profit:", gross_profit)
print("Gross Profit Margin:", gross_profit_margin, "%")
# Interpretation
print("\nInterpretation:")
print(
    "A 40% gross margin means that for every ₹100 of revenue, "
    "the company retains ₹40 after covering COGS.")
print(
    "Management can use gross margin to evaluate pricing, "
    "product costs, and the profitability of core operations.")
```

```python
# Q3 — Operating Profit + Operating Margin
# A company has: Revenue = ₹2000000, COGS = ₹1200000, Operating Expenses = ₹400000
# Calculate: Gross Profit, Operating Profit, Gross Profit Margin, Operating Profit Margin
revenue = 2000000
cogs = 1200000
operating_expenses = 400000
# Gross Profit
gross_profit = revenue - cogs
# Operating Profit
operating_profit = gross_profit - operating_expenses
# Gross Profit Margin
gross_profit_margin = (gross_profit / revenue) * 100
# Operating Profit Margin
operating_profit_margin = (operating_profit / revenue) * 100
print("Gross Profit:", gross_profit)
print("Operating Profit:", operating_profit)
print("Gross Profit Margin:", gross_profit_margin, "%")
print("Operating Profit Margin:", operating_profit_margin, "%")
```

```python
# Q4 — Net Profit + Profit Margin
# A company has: Revenue = ₹2500000, COGS = ₹1400000, Operating Expenses = ₹500000, Interest & Taxes = ₹200000
# Calculate: Gross Profit, Operating Profit, Net Profit, Profit Margin
# Interpret: Is the company keeping a large or small portion of its revenue as final profit?
revenue = 2500000
cogs = 1400000
operating_expenses = 500000
interest_and_taxes = 200000
# Gross Profit
gross_profit = revenue - cogs
# Operating Profit
operating_profit = gross_profit - operating_expenses
# Net Profit
net_profit = operating_profit - interest_and_taxes
# Profit Margin
profit_margin = (net_profit / revenue) * 100
print("Gross Profit:", gross_profit)
print("Operating Profit:", operating_profit)
print("Net Profit:", net_profit)
print("Profit Margin:", profit_margin, "%")
# Interpretation
print("\nInterpretation:")
print(
    "The company keeps 16% of its revenue as final profit "
    "after covering COGS, operating expenses, interest, and taxes.")
if profit_margin >= 20:
    print("The company is keeping a relatively large portion of its revenue as profit.")
else:
    print("The company is keeping a relatively small portion of its revenue as profit.")
```

```python
# Q5 — Break-even Analysis
# A company sells a product for ₹800 per unit. Fixed Costs = ₹400000, Variable Cost per Unit = ₹500
# Calculate:
# 1. Contribution per Unit
# 2. Break-even Point in Units
# 3. Revenue at Break-even
# 4. If the company sells 2,000 units, will it make a profit or loss?
# 5. Calculate the profit/loss.
selling_price = 800
fixed_cost = 400000
variable_cost_per_unit = 500
units_sold = 2000
# 1. Contribution per Unit
contribution_per_unit = selling_price - variable_cost_per_unit
# 2. Break-even Point in Units
break_even_units = fixed_cost / contribution_per_unit
# 3. Revenue at Break-even
break_even_revenue = break_even_units * selling_price
# Revenue and total cost at 2000 units
revenue = units_sold * selling_price
variable_cost = units_sold * variable_cost_per_unit
total_cost = fixed_cost + variable_cost
# 4 & 5. Profit/Loss
profit_loss = revenue - total_cost
print("Contribution per Unit: ₹", contribution_per_unit)
print("Break-even Point:", break_even_units, "units")
print("Revenue at Break-even: ₹", break_even_revenue)
print("Revenue at 2000 units: ₹", revenue)
print("Total Cost at 2000 units: ₹", total_cost)
if profit_loss > 0:
    print("Result: Profit")
elif profit_loss < 0:
    print("Result: Loss")
else:
    print("Result: Break-even")
print("Profit/Loss: ₹", profit_loss)
# Interpretation
print("\nInterpretation:")
print(
    "The company earns ₹300 contribution per unit after covering "
    "the variable cost.")
print(
    "The company must sell approximately 1334 units to cover "
    "its fixed costs and reach break-even.")
print(
    "Since the company sells 2000 units, which is above the "
    "break-even point, it makes a profit.")
```
