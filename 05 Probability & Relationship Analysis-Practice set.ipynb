```python
# BASIC PROBABILITY
print("Basic Probability")
# A company has 500 customers. 320 are active customers. What is the probability that a randomly selected customer is active?
total_customers = 500
active_customers = 320
prob_active = active_customers / total_customers

print("Probability of active customer:", prob_active)
print("Percentage:", prob_active * 100, "%")
```

```python
# PROBABILITY
print("Probability")
# Box contains: Blue = 8, Red = 5, Green = 7
blue = 8
red = 5
green = 7
total_products = blue + red + green
# Probability of red
prob_red = red / total_products
# Probability of blue or green
prob_blue_green = (blue + green) / total_products
#Probability of not red
prob_not_red = (total_products - red) / total_products

print("a) P(Red):", prob_red)
print("b) P(Blue or Green):", prob_blue_green)
print("c) P(Not Red):", prob_not_red)
```

```python
# CONDITIONAL PROBABILITY
print("Conditional Probability")
# Out of 1000 customers: 600 use the mobile app, 240 use the mobile app and make monthly purchases. Find P(Monthly Purchase | Mobile App User)
total_app_users = 600
app_users_monthly_purchase = 240

prob_purchase_given_app = (
    app_users_monthly_purchase / total_app_users)
print(
    "P(Monthly Purchase | Mobile App User):",
    prob_purchase_given_app)
print("Percentage:", prob_purchase_given_app * 100, "%")
```

```python
print("Conditional Probability")
# A company has 1000 customers: 400 are premium customers, 300 are premium customers who renewed. Find P(Renewal | Premium Customer)
premium_customers = 400
premium_renewed = 300

prob_renewal_given_premium = (
    premium_renewed / premium_customers)
print(
    "P(Renewal | Premium Customer):",
    prob_renewal_given_premium)
print("Percentage:", prob_renewal_given_premium * 100, "%")
```

```python
# INDEPENDENT VS DEPENDENT EVENTS
print("Independent vs Dependent Events")
# a) Tossing a coin twice
# b) Drawing two cards without replacement
# c) Today's sales and tomorrow's sales
# d) Rolling a die and tossing a coin
print("a) Tossing a coin twice")
print("Type: Independent")
print("Reason: The first toss does not affect the second toss.")

print("\nb) Drawing two cards without replacement")
print("Type: Dependent")
print("Reason: The first card changes the number of cards available.")

print("\nc) Today's sales and tomorrow's sales")
print("Type: Dependent")
print("Reason: Today's sales can influence or relate to tomorrow's sales.")

print("\nd) Rolling a die and tossing a coin")
print("Type: Independent")
print("Reason: The die outcome does not affect the coin outcome.")
```

```python
# EXPECTED VALUE
print("Expected Value")
# Marketing campaign outcomes:
# High sales:   Probability = 0.30, Profit = ₹50000
# Medium sales: Probability = 0.50, Profit = ₹20000
# Low sales:    Probability = 0.20, Profit = -₹10000
high_sales_probability = 0.30
medium_sales_probability = 0.50
low_sales_probability = 0.20

high_sales_profit = 50000
medium_sales_profit = 20000
low_sales_profit = -10000

expected_profit = (
    high_sales_probability * high_sales_profit
    + medium_sales_probability * medium_sales_profit
    + low_sales_probability * low_sales_profit)
print("Expected Profit: ₹", expected_profit)
```

```python
# EXPECTED VALUE
print("Expected Value")
# Investment:
# 40% probability of earning ₹10000
# 35% probability of earning ₹4000
# 25% probability of losing ₹6000
probability_1 = 0.40
probability_2 = 0.35
probability_3 = 0.25

return_1 = 10000
return_2 = 4000
return_3 = -6000

expected_return = (
    probability_1 * return_1
    + probability_2 * return_2
    + probability_3 * return_3)
print("Expected Return: ₹", expected_return)
if expected_return > 0:
    print("Expected Value: Positive")
else:
    print("Expected Value: Negative")
```

```python
# COVARIANCE
print("Covariance")
import numpy as np
# Advertising and Sales
advertising = np.array([10, 20, 30, 40, 50])
sales = np.array([50, 60, 70, 80, 90])

covariance = np.cov(advertising, sales, ddof=1)[0, 1]
print("Covariance:", covariance)
if covariance > 0:
    print("a) Covariance is Positive")
else:
    print("a) Covariance is Negative")
print("b) Reason: Advertising and sales move in the same direction.")
```

```python
# COVARIANCE INTERPRETATION
print("Covariance Interpretation")
# Cov(X,Y) = -125
covariance = -125
print("Cov(X,Y):", covariance)
print(
    "Interpretation: The negative sign indicates that X and Y "
    "tend to move in opposite directions.")
print(
    "Covariance alone does not tell us whether the relationship "
    "is strong or weak.")
print(
    "Reason: Covariance depends on the units and scale of the variables.")
```

```python
# CORRELATION
print("Correlation")
# Interpret:
# a) r = +0.91
# b) r = -0.82
# c) r = +0.08
# d) r = -0.05
correlations = {
    "a) r = +0.91": 0.91,
    "b) r = -0.82": -0.82,
    "c) r = +0.08": 0.08,
    "d) r = -0.05": -0.05}
for label, r in correlations.items():
    if abs(r) >= 0.70:
        strength = "Strong"
    elif abs(r) >= 0.30:
        strength = "Moderate"
    else:
        strength = "Weak"
    if r > 0:
        direction = "Positive"
    else:
        direction = "Negative"
    print(label)
    print("Strength:", strength)
    print("Direction:", direction)
    print()
```

```python
# CORRELATION COMPARISON
print("Correlation Comparison")
# Relationship A: r = +0.72
# Relationship B: r = -0.88
r_A = 0.72
r_B = -0.88

print("Relationship A:", r_A)
print("Relationship B:", r_B)

if abs(r_A) > abs(r_B):
    print("Stronger relationship: A")
else:
    print("Stronger relationship: B")
print("Reason: Correlation strength is based on the absolute value of r.")
```

```python
# PEARSON CORRELATION
print("Pearson Correlation")
from scipy.stats import pearsonr
# Advertising and Sales
advertising = np.array([10, 20, 30, 40, 50])
sales = np.array([20, 30, 40, 50, 60])

pearson_r, p_value = pearsonr(advertising, sales)
print("Pearson Correlation:", pearson_r)
print(
    "Interpretation: There is a very strong positive linear "
    "relationship between advertising and sales.")
```

```python
# CORRELATION ≠ CAUSATION
print("Correlation ≠ Causation")
# Training attendance and employee performance: Correlation = +0.78
correlation = 0.78
print("Correlation:", correlation)

print("Conclusion: Management cannot conclude that training causes higher employee performance.")
print("Reason: Correlation shows association, not causation.")
print("Possible third variable: Employee motivation.")
print("Motivated employees may attend more training and also perform better.")
```

```python
# BUSINESS INTERPRETATION
print("Business Interpretation")
# Correlation between discount percentage and units sold = +0.67
correlation = 0.67

print("Correlation between discount percentage and units sold:", correlation)
print("Interpretation: There is a moderately strong positive relationship between discount percentage and units sold.")
print("This suggests that higher discounts are associated with higher unit sales.")
print("It does NOT prove that discounts cause higher sales.")
```

```python
# IDENTIFY VARIABLES
print("Identify Variables")
# A company wants to determine whether advertising expenditure predicts sales revenue.
print("a) Independent Variable: Advertising Expenditure")
print("b) Dependent Variable: Sales Revenue")
print("c) Predictor: Advertising Expenditure")
print("d) Outcome: Sales Revenue")
```

```python
# REGRESSION EQUATION
print("Regression Equation")
# Sales = 25,000 + 3.5 × Advertising. Advertising is measured in ₹000.
intercept = 25000
slope = 3.5

print("Regression Equation:")
print("Sales = 25,000 + 3.5 × Advertising")

print("a) Intercept:", intercept)
print("b) Slope:", slope)
print("c) Slope Meaning: For every additional ₹1,000 spent on advertising, predicted sales increase by ₹3,500.")
print("d) Intercept Meaning: When advertising expenditure is zero, predicted sales are ₹25,000.")
```

```python
# PREDICTION
print("Prediction")
# Sales = 25000 + 3.5 × Advertising. Advertising is measured in ₹000.
# Advertising = ₹10000 = 10 in ₹000
advertising = 10
predicted_sales = 25000 + (3.5 * advertising * 1000)

print("Advertising Expenditure: ₹10,000")
print("Predicted Sales: ₹", predicted_sales)
```

```python
# PREDICTION
print("Prediction")
# Sales = 25000 + 3.5 × Advertising. Advertising is measured in ₹000.
advertising_20 = 20
advertising_30 = 30
advertising_50 = 50

sales_20 = 25000 + (3.5 * advertising_20 * 1000)
sales_30 = 25000 + (3.5 * advertising_30 * 1000)
sales_50 = 25000 + (3.5 * advertising_50 * 1000)

print("a) Advertising = ₹20,000")
print("   Predicted Sales: ₹", sales_20)

print("b) Advertising = ₹30,000")
print("   Predicted Sales: ₹", sales_30)

print("c) Advertising = ₹50,000")
print("   Predicted Sales: ₹", sales_50)
```

```python
# INTERPRET THE SLOPE
print("Interpret the Slope")
# Profit = 12000 + 2.8 × Advertising. Advertising is measured in ₹000.
slope = 2.8

print("Slope:", slope)
print("Interpretation: For every additional ₹1,000 spent on advertising, predicted profit increases by ₹2,800.")
```

```python
# INTERPRET THE INTERCEPT
print("Interpret the Intercept")
# Sales = 40000 + 5 × Advertising
intercept = 40000

print("Intercept:", intercept)
print("Interpretation: When advertising expenditure is zero, predicted sales are ₹40000.")
print("The intercept may not always be realistic to interpret literally if zero advertising is outside the observed business range.")
```

```python
# R² INTERPRETATION
print("R² Interpretation")

r_squared = 0.64

print("R²:", r_squared)
print("Percentage Explained:", r_squared * 100, "%")
print("Interpretation: 64% of the variation in the dependent variable is explained by the regression model.")
```

```python
# COMPARE REGRESSION MODELS
print("Compare Models")
model_A = 0.42
model_B = 0.71
model_C = 0.28

print("Model A R²:", model_A)
print("Model B R²:", model_B)
print("Model C R²:", model_C)

best_model = max(
    {"Model A": model_A, "Model B": model_B, "Model C": model_C},
    key={"Model A": model_A, "Model B": model_B, "Model C": model_C}.get)
best_r_squared = max(model_A, model_B, model_C)

print("Best Model:", best_model)
print("Variation Explained:", best_r_squared * 100, "%")
```

```python
# REGRESSION INTERPRETATION
print("Regression Interpretation")
correlation = 0.80
r_squared = 0.64
slope = 4.5

print("\nCorrelation:", correlation)
print("Interpretation: There is a strong positive relationship between the two variables.")

print("\nR²:", r_squared)
print("Interpretation: 64% of the variation in the dependent variable is explained by the model.")

print("\nSlope:", slope)
print("Interpretation: For every one-unit increase in the independent variable, predicted dependent variable increases by 4.5 units.")
```

```python
# FULL BUSINESS CASE
print("Full Business Case")
# Sales = 80000 + 4 × Advertising. Both variables are measured in ₹000. R² = 0.72
intercept = 80000
slope = 4
r_squared = 0.72
advertising = 25
predicted_sales = (
    intercept
    + slope * advertising * 1000)
print("a) Independent Variable: Advertising Expenditure")
print("b) Dependent Variable: Sales")
print("c) Slope:", slope)
print("d) Slope Interpretation: For every additional ₹1000 spent on advertising, predicted sales increase by ₹4000.")
print("e) Intercept:", intercept)
print("f) Intercept Interpretation: When advertising expenditure is zero, predicted sales are ₹80000.")
print("g) R²:", r_squared)
print("R² Interpretation: 72% of the variation in sales is explained by advertising in this regression model.")
print("h) Advertising Expenditure: ₹25000")
print("Predicted Sales: ₹", predicted_sales)
print("i) Advertising has a meaningful relationship with sales because the model explains 72% of the variation in sales.")
print("However, this does not prove that advertising causes sales to increase.")
```

```python
# MANAGEMENT DECISION
print("Management Decision")
# Analysis: Correlation = 0.86, Regression slope = 3.2, R² = 0.74
correlation = 0.86
slope = 3.2
r_squared = 0.74

print("Correlation:", correlation)
print("Regression Slope:", slope)
print("R²:", r_squared)

print("\nManagement's Conclusion:")
print("Advertising causes 74% of our sales, so we should double the advertising budget.")

print("\nProblems with this conclusion:")

print("1. Correlation does not prove causation.")

print("2. R² = 0.74 means 74% of the variation in sales is explained by the model, not that advertising causes 74% of sales.")

print("3. Doubling advertising does not guarantee that sales will double.")
print("\nBefore increasing the advertising budget, management should investigate:")

print("- Return on Advertising Spend (ROAS)")
print("- Marginal sales from additional advertising")
print("- Seasonality")
print("- Pricing and discounts")
print("- Competitor activity")
print("- Market demand")
print("- Controlled experiments / A-B testing")
```

```python
# FINAL INTEGRATED QUESTION
print("Final Integrated Question")
from scipy.stats import pearsonr, linregress
# Data
months = np.array([1, 2, 3, 4, 5, 6])
advertising = np.array([10, 15, 20, 25, 30, 35])
sales = np.array([55, 62, 68, 75, 81, 90])
# Pearson Correlation
correlation, p_value = pearsonr(advertising, sales)
# Covariance
covariance = np.cov(advertising, sales, ddof=1)[0, 1]
# Linear Regression
regression = linregress(advertising, sales)
slope = regression.slope
intercept = regression.intercept
r_squared = regression.rvalue ** 2
# a) Relationship
print("a) Expected Relationship:")
print("Positive relationship — as advertising increases, sales also increase.")
# b) Direction of correlation
print("\nb) Correlation Direction:")

if correlation > 0:
    print("Positive")
else:
    print("Negative")
# c) Strength
print("\nc) Correlation Strength:")
print("Pearson Correlation:", correlation)

if abs(correlation) >= 0.70:
    print("Strong")
elif abs(correlation) >= 0.30:
    print("Moderate")
else:
    print("Weak")
# d) Covariance
print("\nd) Covariance:")
print("Covariance:", covariance)

if covariance > 0:
    print("Positive")
else:
    print("Negative")
# e) Independent Variable
print("\ne) Independent Variable:")
print("Advertising Expenditure")
# f) Dependent Variable
print("\nf) Dependent Variable:")
print("Sales")
# g) Regression Line
print("\ng) Purpose of Regression Line:")
print("The regression line is used to predict sales based on advertising expenditure.")
# h) R²
print("\nh) R²:")
print("R²:", r_squared)

print("Percentage Explained:", r_squared * 100, "%")
print(
    "Interpretation: Approximately",
    round(r_squared * 100, 2),
    "% of the variation in sales is explained by advertising in this model.")
# i) Causation
print("\ni) Does advertising cause sales to increase?")
print("No.")

print("A strong correlation and high R² do not by themselves prove causation.")
# j) Management Recommendation
print("\nj) Management Recommendation:")
print(
    "Management should investigate the relationship further using controlled experiments, ROI analysis and other variables"
    "such as seasonality, pricing, promotions, competitor activity and market demand before making a major advertising-budget decision.")
```
