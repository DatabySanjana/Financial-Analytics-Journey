#### Part A - FP&A Fundamentals

##### Q1. FP&A Role
1. What is FP&A?

 - FP&A stands for Financial Planning and Analysis. It is a function in a company that deals with financial planning, budgeting, forecasting and analysing business performance. It helps management understand the financial position of the business and make better decisions about the future.

2. What does an FP&A Analyst do?

 - An FP&A Analyst works with financial data to prepare budgets and forecasts and compare actual results with planned results. They also analyse variances and try to understand the reasons behind them. The analyst may also prepare reports, track KPIs and provide useful information to management for decision-making.

3. How is FP&A different from Accounting?

 - Accounting mainly focuses on recording and reporting financial transactions that have already happened. FP&A focuses more on analysing this information and using it for budgeting, forecasting and future planning. In simple terms, accounting helps understand what happened, while FP&A helps understand why it happened and what may happen in the future.

##### Q2. Budget vs Forecast vs Actuals
Explain the difference between **Budget, Forecast and Actuals**.

Answer:

**Budget:** A budget is a financial plan prepared for a future period. It estimates expected revenue, expenses, profit and other financial activities based on the company's goals and assumptions.

**Forecast:** A forecast is an updated estimate of what the company expects to happen in the future. It is usually revised using the latest actual results, business conditions and available information.

**Actuals:** Actuals are the financial results that have actually occurred during a particular period. They are recorded based on the company's real revenue, expenses and other financial transactions. start the calculations.


### Part B - Budgeting

```python
# Revenue Budget
# Q3. A company expects to sell: January: 10000 units at ₹500 per unit, February: 12000 units at ₹500 per unit, March: 14000 units at ₹520 per unit
# Calculate: Budgeted revenue for each month & Total budgeted revenue
import pandas as pd
data = {
    "Month": ["January", "February", "March"],
    "Units": [10000, 12000, 14000],
    "Price": [500, 500, 520]}
df = pd.DataFrame(data)
df["Revenue"] = df["Units"] * df["Price"]

print(df)
print("Total Budgeted Revenue:", df["Revenue"].sum())
print("\nInterpretation: March has the highest budgeted revenue because both expected sales volume and selling price are higher.")
```

```python
# Expense Budget
# Q4. A department has the following expenses: Salaries = ₹500000, Rent = ₹120000, Marketing = ₹80000, Utilities = ₹40000, Other expenses = ₹60000
# Calculate: Total budgeted expenses & Each expense as % of total expenses
import pandas as pd
data = {
    "Expense": ["Salaries", "Rent", "Marketing", "Utilities", "Other Expenses"],
    "Amount": [500000, 120000, 80000, 40000, 60000]}
df = pd.DataFrame(data)
# Calculate total expenses
total_expenses = df["Amount"].sum()
# Calculate percentage of total expenses
df["Percentage"] = (df["Amount"] / total_expenses) * 100

print(df)
print("\nTotal Budgeted Expenses:", total_expenses)
print("\nInterpretation: Salaries are the largest expense, accounting for 62.5% of total departmental expenses. Therefore, changes in salary costs would have the biggest impact on the department's overall expense budget.")
```

```python
# Profit Budget
# Q5. Using the total revenue from Q3 and total expenses from Q4: Calculate Budgeted Profit & Calculate Profit Margin
# Formula:
# Budgeted Profit = Revenue - Expenses
# Profit Margin = (Profit / Revenue) * 100
total_revenue = 18280000
total_expenses = 800000
# Calculate budgeted profit
budgeted_profit = total_revenue - total_expenses
# Calculate profit margin
profit_margin = (budgeted_profit / total_revenue) * 100

print("Budgeted Profit:", budgeted_profit)
print("Profit Margin:", round(profit_margin, 2), "%")
print("\nInterpretation: The calculation is only an illustration using the given figures; in a real FP&A analysis, all relevant company expenses"
      "would need to be included before calculating the actual profit margin.")
```

Top-down vs Bottom-up Budgeting

Q6. A company prepares its budget in two different ways:

Company A: Management sets a ₹50 crore annual expense target and distributes it among departments.
Company B: Each department prepares its own expense plan and these plans are combined to create the overall company budget.

Identify which is **Top-down** and which is **Bottom-up**, and give one advantage of each.

**Company A - Top-down budgeting:**
The budget is decided by senior management and then allocated to different departments.

**Advantage:** It ensures that departmental budgets are aligned with the company's overall financial goals.

**Company B - Bottom-up budgeting:**
Individual departments prepare their own budgets, which are then combined into the overall company budget.

**Advantage:** Department managers usually have better knowledge of their operational requirements, making the budget more realistic.

**Interpretation**: Top-down focuses on management targets, while bottom-up focuses on department-level estimates and participation.

### Part C - Variance Analysis

```python
# Budget vs Actual
# Q7. Calculate: Revenue variance, Expense variance, Budgeted profit, Actual profit, Profit variance.
# Classify each variance as Favorable or Unfavorable.
budget_revenue = 1000000
actual_revenue = 900000

budget_expenses = 600000
actual_expenses = 650000
# Calculate variances
revenue_variance = actual_revenue - budget_revenue
expense_variance = actual_expenses - budget_expenses
# Calculate profit
budgeted_profit = budget_revenue - budget_expenses
actual_profit = actual_revenue - actual_expenses
# Calculate profit variance
profit_variance = actual_profit - budgeted_profit

print("Revenue Variance:", revenue_variance)
print("Expense Variance:", expense_variance)
print("Budgeted Profit:", budgeted_profit)
print("Actual Profit:", actual_profit)
print("Profit Variance:", profit_variance)

print("\nInterpretation: Lower revenue and higher expenses reduced actual profit by ₹1.5 lakh compared with the budget.")
```

```python
# Variance %
# Q8. Using Q7, calculate: Revenue variance %, Expense variance %, Profit variance %
# Formula: Variance % = ((Actual - Budget) / Budget) * 100
budget_revenue = 1000000
actual_revenue = 900000

budget_expenses = 600000
actual_expenses = 650000

budgeted_profit = budget_revenue - budget_expenses
actual_profit = actual_revenue - actual_expenses
# Calculate variance percentages
revenue_variance_pct = ((actual_revenue - budget_revenue) / budget_revenue) * 100
expense_variance_pct = ((actual_expenses - budget_expenses) / budget_expenses) * 100
profit_variance_pct = ((actual_profit - budgeted_profit) / budgeted_profit) * 100

print("Revenue Variance %:", revenue_variance_pct)
print("Expense Variance %:", expense_variance_pct)
print("Profit Variance %:", profit_variance_pct)

print("\nInterpretation: Revenue fell by 10% and expenses increased by 8.33%, resulting in a 37.5% decline in profit against budget.")
```

**Revenue Variance Explanation**

Q9. A company's revenue was **₹20 lakh below budget**, but management reports that:
 - Sales volume was **10% lower than expected**
 - Selling price was **5% higher than budget**
 - Explain how **volume variance** and **price variance** could have contributed to the overall revenue variance.

Revenue depends on both **quantity sold and selling price**:

**Revenue = Sales Volume × Selling Price**

The **10% decrease in sales volume** would reduce revenue because fewer units were sold than expected. On the other hand, the **5% increase in selling price** would increase revenue because each unit generated more revenue.

Therefore, the higher selling price partially offset the negative impact of the lower sales volume, but the volume decline was large enough to result in an overall **₹20 lakh unfavorable revenue variance**.

**Interpretation**: The main negative factor was lower sales volume, while the higher selling price provided a partial offset.: **Q10 — Cost Variance.**


```python
# Cost Variance
# Q10. Calculate: Budgeted cost, Actual cost, Total cost variance, Explain the reason for the variance
# Budget: Production = 10000 units & Cost per unit = ₹200
# Actual: Production = 12000 units & Cost per unit = ₹220
budget_units = 10000
budget_cost_per_unit = 200

actual_units = 12000
actual_cost_per_unit = 220
# Calculate total costs
budgeted_cost = budget_units * budget_cost_per_unit
actual_cost = actual_units * actual_cost_per_unit
# Calculate cost variance
cost_variance = actual_cost - budgeted_cost

print("Budgeted Cost:", budgeted_cost)
print("Actual Cost:", actual_cost)
print("Total Cost Variance:", cost_variance)

print("\nInterpretation: Higher production volume and higher unit cost together caused an unfavorable cost variance of ₹6.4 lakh.")
```

#### Part D - Forecasting

```python
# Growth-Based Forecast
# Q11. Historical Revenue: 2023 = ₹10 Cr, 2024 = ₹12 Cr, 2025 = ₹15 Cr,2026 = ₹18 Cr
# Calculate: YoY growth for each year, Average growth rate, Forecast 2027 revenue using average growth rate
data = {
    "Year": [2023, 2024, 2025, 2026],
    "Revenue": [10, 12, 15, 18]}
df = pd.DataFrame(data)
# Calculate YoY growth
df["YoY_Growth_%"] = df["Revenue"].pct_change() * 100
# Calculate average growth rate
average_growth = df["YoY_Growth_%"].dropna().mean()
# Forecast 2027 revenue
forecast_2027 = df["Revenue"].iloc[-1] * (1 + average_growth / 100)

print(df)
print("\nAverage Growth Rate:", round(average_growth, 2), "%")
print("Forecast Revenue for 2027:", round(forecast_2027, 2), "Cr")

print("\nInterpretation: Based on the average historical growth rate of 21.67%, 2027 revenue is forecast at approximately ₹21.9 crore.")
```

```python
# Moving Average
# Q12. Monthly revenue: 100, 110, 120, 130, 140, 150
# Calculate the 3-month moving average forecast for the next month.
revenue = [100, 110, 120, 130, 140, 150]
df = pd.DataFrame({
    "Month": range(1, 7),
    "Revenue": revenue})
# Calculate 3-month moving average
df["3_Month_MA"] = df["Revenue"].rolling(window=3).mean()
# Forecast next month using the latest 3 months
next_month_forecast = df["Revenue"].tail(3).mean()

print(df)
print("\nNext Month Forecast:", next_month_forecast)
```

```python
# Trend Forecasting
# Q13. Revenue has increased by approximately ₹5 lakh every year.
# Calculate the forecast revenue for 2027 using the existing revenue trend.
data = {
    "Year": [2022, 2023, 2024, 2025, 2026],
    "Revenue": [20, 25, 30, 35, 40]}
df = pd.DataFrame(data)
# Calculate yearly change in revenue
df["Revenue_Change"] = df["Revenue"].diff()
# Calculate average yearly increase
average_increase = df["Revenue_Change"].dropna().mean()
# Forecast 2027 revenue
forecast_2027 = df["Revenue"].iloc[-1] + average_increase

print(df)
print("\nAverage Yearly Increase:", average_increase, "Lakh")
print("Forecast Revenue for 2027:", forecast_2027, "Lakh")
```

### Part E - Scenario Analysis

```python
# Three-Scenario Analysis
# Q14. Current Revenue = ₹100 Cr, Best Case = +15%, Base Case = +8%, Worst Case = -5%
# Calculate revenue under each scenario.
current_revenue = 100
# Calculate scenario revenues
best_case = current_revenue * (1 + 0.15)
base_case = current_revenue * (1 + 0.08)
worst_case = current_revenue * (1 - 0.05)

print("Best Case Revenue:", best_case, "Cr")
print("Base Case Revenue:", base_case, "Cr")
print("Worst Case Revenue:", worst_case, "Cr")
print("\nInterpretation: The scenario analysis shows that revenue could range from ₹95 Cr to ₹115 Cr depending on business conditions.")
```

```python
# Sensitivity Analysis
# Q15. Sales volume = 100000 units, Selling price = ₹500 per unit, Variable cost = ₹300 per unit, Fixed cost = ₹10000000
# Calculate: Revenue, Variable cost, Contribution, Profit, Effect of a 10% decrease in selling price
sales_volume = 100000
selling_price = 500
variable_cost_per_unit = 300
fixed_cost = 10000000
# Original scenario
revenue = sales_volume * selling_price
variable_cost = sales_volume * variable_cost_per_unit
contribution = revenue - variable_cost
profit = contribution - fixed_cost
# New selling price after 10% decrease
new_selling_price = selling_price * 0.90
# New scenario
new_revenue = sales_volume * new_selling_price
new_contribution = new_revenue - variable_cost
new_profit = new_contribution - fixed_cost

print("Revenue:", revenue)
print("Variable Cost:", variable_cost)
print("Contribution:", contribution)
print("Profit:", profit)

print("\nNew Selling Price:", new_selling_price)
print("New Revenue:", new_revenue)
print("New Profit:", new_profit)
print("Change in Profit:", new_profit - profit)

print("\nInterpretation: A 10% price reduction cuts profit by 50%, showing that profitability is highly sensitive to selling price.")
```

### Part F - Business KPIs

```python
# Business KPI Calculations
# Q16. Calculate: Gross Margin, Operating Margin, Average Revenue per Customer, CAC, Churn Rate
# Revenue = ₹50 Cr, Gross Profit = ₹20 Cr, Operating Profit = ₹8 Cr, Customers = 100000, Marketing & Sales Cost = ₹5 Cr, Lost Customers = 8000, Starting Customers = 100,000
revenue = 50       # ₹ Cr
gross_profit = 20  # ₹ Cr
operating_profit = 8  # ₹ Cr
customers = 100000

marketing_sales_cost = 5  # ₹ Cr
lost_customers = 8000
starting_customers = 100000
# Gross Margin
gross_margin = (gross_profit / revenue) * 100
# Operating Margin
operating_margin = (operating_profit / revenue) * 100
# Average Revenue per Customer
# Convert ₹ Cr to ₹
average_revenue_per_customer = (revenue * 10_000_000) / customers
# Customer Acquisition Cost
cac = (marketing_sales_cost * 10_000_000) / customers
# Churn Rate
churn_rate = (lost_customers / starting_customers) * 100

print("Gross Margin:", gross_margin, "%")
print("Operating Margin:", operating_margin, "%")
print("Average Revenue per Customer: ₹", average_revenue_per_customer)
print("CAC: ₹", cac)
print("Churn Rate:", churn_rate, "%")

print("\nInterpretation: The company has a 40% gross margin and 16% operating margin, while 8% customer churn indicates some customer retention pressure.")
```

```python
# EBITDA Concept
# Q17. Calculate EBITDA and explain its usefulness.
# Revenue = ₹100 Cr, Operating Expenses = ₹60 Cr, Depreciation = ₹5 Cr, Amortization = ₹2 Cr
revenue = 100  # ₹ Cr
operating_expenses = 60  # ₹ Cr
depreciation = 5  # ₹ Cr
amortization = 2  # ₹ Cr
# EBITDA = Revenue - Operating Expenses
# excluding depreciation and amortization
ebitda = revenue - operating_expenses

print("EBITDA:", ebitda, "Cr")

print("\nInterpretation: The company generated ₹40 crore of EBITDA, indicating its operating earnings before interest, taxes, depreciation and amortization.")

print("\nWhy is EBITDA useful?")
print("EBITDA focuses on the company's operating performance before interest, taxes, depreciation and amortization. It can help management compare"
      "operating performance across periods or between companies without the effect of financing and certain accounting decisions.")
```

##### Final FP&A Interpretation

Q20. A company reports: Revenue: +10% vs budget, Expenses: +15% vs budget, Profit: −5% vs budget, Customer churn: +3 percentage points, CAC: +12%

As an FP&A Analyst, identify: Key findings, Possible causes, Business risks, Recommended management actions

Answer:

1. Key Findings:

The company generated **10% higher revenue than budget**, which is positive. However, expenses increased by **15%**, which is higher than the revenue growth and resulted in profit being **5% below budget**. At the same time, customer churn increased by 3 percentage points and CAC increased by 12%.

2. Possible Causes:

The higher expenses may be due to increased operating costs, marketing spending or employee costs. Higher CAC may indicate that acquiring new customers has become more expensive, while the increase in churn suggests that the company is losing more existing customers.

3. Business Risks:

The main risks are **declining profitability, rising customer acquisition costs and customer retention problems**. If these trends continue, the company may face further pressure on margins and sustainable growth.

4. Recommended Management Actions:

Management should analyse the major expense variances and identify areas where costs can be controlled. The company should also investigate the reasons for higher churn and CAC and evaluate the effectiveness of its marketing and customer retention strategies. The forecast should then be updated based on these findings.

Interpretation: Although revenue is above budget, rising costs, CAC and churn are creating profitability and long-term growth concerns.
